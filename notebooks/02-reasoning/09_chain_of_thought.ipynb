## 📋 Description

Chain-of-Thought (CoT) prompting is a technique that encourages language models to break down complex problems into intermediate reasoning steps before providing a final answer. By explicitly showing the model how to think through a problem step-by-step, CoT significantly improves performance on tasks requiring arithmetic, commonsense, and symbolic reasoning.

**When to use:**
- Multi-step math problems
- Logical reasoning tasks
- Complex decision-making scenarios
- Problems requiring multiple inferences

## 🔧 How It Works

```
Standard Prompting:                    Chain-of-Thought Prompting:
┌─────────────────┐                    ┌─────────────────┐
│ Input:          │                    │ Input:          │
│ Q: 23 + 47 = ?  │                    │ Q: 23 + 47 = ?  │
│                 │                    │ Let's solve     │
│ Output:         │                    │ step-by-step:   │
│ 70              │                    │                 │
└─────────────────┘                    │ Step 1: 23 + 40 │
                                       │         = 63    │
                                       │ Step 2: 63 + 7  │
                                       │         = 70    │
                                       │                 │
                                       │ Answer: 70      │
                                       └─────────────────┘
```

**Key Steps:**
1. Present the problem clearly
2. Add reasoning steps in the prompt (few-shot examples)
3. Show intermediate calculations
4. Arrive at the final answer

## 💡 Basic Example

Let's see the difference between standard prompting and Chain-of-Thought for a math word problem.

**Problem:** A bakery made 48 cupcakes. They sold 15 in the morning and 20 in the afternoon. How many cupcakes are left?

````python
# Standard Prompting (without CoT)standard_prompt = """Q: A bakery made 48 cupcakes. They sold 15 in the morning and 20 in the afternoon. How many cupcakes are left?A: 13Q: A store had 100 apples. They sold 30 on Monday and 25 on Tuesday. How many apples are left?A:"""print("="*50)print("❌ Standard Prompting (no reasoning):")print("="*50)response1 = get_completion(standard_prompt, temperature=0)print(response1)# Chain-of-Thought Promptingprint("" + "="*50)print("✅ Chain-of-Thought Prompting:")print("="*50)cot_prompt = """Q: A bakery made 48 cupcakes. They sold 15 in the morning and 20 in the afternoon. How many cupcakes are left?A: Let's think step by step. First, find total sold: 15 + 20 = 35 cupcakes. Then subtract from total: 48 - 35 = 13 cupcakes left. The answer is 13.Q: A store had 100 apples. They sold 30 on Monday and 25 on Tuesday. How many apples are left?A:"""response2 = get_completion(cot_prompt, temperature=0)print(response2)
````

## ⚠️ Failure Case

CoT can fail when:
1. The reasoning steps are incorrect or misleading
2. The model hallucinates intermediate steps
3. Too many steps cause the model to lose track
4. The problem requires external knowledge not in the prompt

````python
# Failure Case: Incorrect reasoning stepsbad_cot_prompt = """Q: If a train travels 60 miles in 1 hour, how far will it travel in 45 minutes?A: Let's think step by step. First, convert 45 minutes to hours: 45/60 = 0.75 hours. Then multiply speed by time: 60 miles/hour * 0.75 hours = 45 miles. The answer is 45 miles.Q: A car travels 80 miles in 2 hours. How far will it travel in 30 minutes?A:"""print("="*60)print("⚠️ Example where CoT works correctly:")print("="*60)response = get_completion(bad_cot_prompt, temperature=0)print(response)# Now show a case where CoT might failprint("" + "="*60)print("⚠️ Case where CoT can fail (ambiguous problem):")print("="*60)ambiguous_prompt = """Q: John has 3 sisters. Each sister has 1 brother. How many brothers does John have?A: Let's think step by step."""response2 = get_completion(ambiguous_prompt, temperature=0.7, model="gpt-3.5-turbo")print(response2)print("💡 Note: This can fail if the model doesn't recognize John IS the brother!")
````

## 💡 Tips & Tricks

**Model-Specific Advice:**
- **GPT-4/Claude-3:** Excellent at following CoT; use 3-5 examples
- **GPT-3.5:** Needs more explicit step-by-step instructions
- **Smaller models:** May need very detailed, simple steps

**Best Practices:**
- ✓ Use clear, consistent formatting for reasoning steps
- ✓ Include the final answer clearly marked
- ✓ Use diverse examples covering different problem types
- ✓ Keep reasoning steps logically connected
- ✗ Don't skip steps - even 'obvious' ones
- ✗ Don't use inconsistent formatting between examples
