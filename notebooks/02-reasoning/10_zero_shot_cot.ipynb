## 📋 Description

Zero-Shot Chain-of-Thought (Zero-Shot CoT) is a technique where you add a simple trigger phrase like "Let's think step by step" to elicit reasoning without providing any examples. This approach requires no manual example crafting and works surprisingly well across many reasoning tasks.

**When to use:**
- When you don't have example reasoning chains
- Quick experiments to test if reasoning helps
- When you want minimal prompt engineering
- As a baseline before trying few-shot CoT

## 🔧 How It Works

```
Standard Zero-Shot:                    Zero-Shot CoT:
┌─────────────────────┐                ┌─────────────────────┐
│ Q: Roger has 5      │                │ Q: Roger has 5      │
│    tennis balls.    │                │    tennis balls.    │
│ He buys 2 more      │                │ He buys 2 more      │
│ cans. Each has 3    │                │ cans. Each has 3    │
│ balls. How many     │                │ balls. How many     │
│ does he have now?   │                │ does he have now?   │
│                     │                │                     │
│ A: 11               │                │ A: Let's think      │
└─────────────────────┘                │    step by step.    │
                                       │    Roger starts     │
                                       │    with 5 balls.    │
                                       │    He buys 2 cans   │
                                       │    = 2 × 3 = 6      │
                                       │    balls.           │
                                       │    Total: 5 + 6     │
                                       │    = 11 balls.      │
                                       └─────────────────────┘
```

**The Magic Phrase:**
- "Let's think step by step"
- "Let's solve this step by step"
- "Let's work through this systematically"
- "Explain your reasoning"

## 💡 Basic Example

Comparing standard zero-shot with Zero-Shot CoT on a simple math problem.

**Problem:** A farmer has 15 sheep and sells 7. Then he buys 12 more. How many sheep does he have?

````python
# Zero-Shot (without reasoning trigger)zero_shot_prompt = """Q: A farmer has 15 sheep and sells 7. Then he buys 12 more. How many sheep does he have?A:"""print("="*50)print("❌ Standard Zero-Shot:")print("="*50)response1 = get_completion(zero_shot_prompt, temperature=0)print(response1)# Zero-Shot CoTprint("" + "="*50)print("✅ Zero-Shot CoT (with 'Let's think step by step'):")print("="*50)zscot_prompt = """Q: A farmer has 15 sheep and sells 7. Then he buys 12 more. How many sheep does he have?A: Let's think step by step."""response2 = get_completion(zscot_prompt, temperature=0)print(response2)
````

## ⚠️ Failure Case

Zero-Shot CoT can fail when:
1. The problem requires domain-specific knowledge
2. The reasoning needs to follow specific formats
3. The model generates incorrect intermediate steps
4. Commonsense reasoning is required beyond pattern matching

````python
# Failure Case: Zero-Shot CoT limitations# Case 1: Commonsense reasoning failurecommonsense_prompt = """Q: A room has 3 lamps. How many light bulbs are needed if each lamp needs 2 bulbs?A: Let's think step by step."""print("="*60)print("⚠️ Simple case (works well):")print("="*60)response1 = get_completion(commonsense_prompt, temperature=0)print(response1)# Case 2: Where it might struggleprint("" + "="*60)print("⚠️ Complex reasoning (may fail):")print("="*60)complex_prompt = """Q: In a game, you have a 60% chance to win each round. What's the probability of winning at least 3 out of 5 rounds?A: Let's think step by step."""response2 = get_completion(complex_prompt, temperature=0)print(response2)print("💡 Note: Statistical problems may need few-shot examples or PAL (Program-Aided LLM)")
````

## 💡 Tips & Tricks

**Effective Trigger Phrases:**
- "Let's think step by step" (most popular)
- "Let's work through this systematically"
- "Let's break this down step by step"
- "Explain your reasoning process"

**Best Practices:**
- ✓ Place the trigger at the end of the prompt
- ✓ Use temperature=0 for consistent reasoning
- ✓ Combine with clear problem statements
- ✓ Try multiple trigger phrases if one doesn't work
- ✗ Don't expect perfection on complex problems
- ✗ Don't use for tasks requiring specific reasoning formats
