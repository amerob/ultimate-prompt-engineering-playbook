## 📋 Description

Decomposed Prompting breaks complex problems into smaller, manageable sub-problems that can be solved independently. Each sub-problem is easier for the model to handle, and the solutions are then combined to solve the original problem.

**When to use:**
- Multi-part questions with distinct components
- Problems requiring different types of reasoning
- Complex analysis with multiple dimensions
- Tasks that benefit from modular thinking

## 🔧 How It Works

```
Decomposed Prompting Process:

┌─────────────────────────────────────────────┐
│         Complex Problem                     │
│  "Analyze company performance"             │
└──────────────────┬──────────────────────────┘
                   │
                   ▼
         ┌─────────────────┐
         │  Decomposer     │
         │  (breaks into   │
         │   sub-problems) │
         └────────┬────────┘
                  │
    ┌─────────────┼─────────────┐
    ▼             ▼             ▼
 ┌──────┐    ┌──────┐    ┌──────┐
 │Sub 1 │    │Sub 2 │    │Sub 3 │
 │Revenue│   │Costs │    │Growth│
 └──┬───┘    └──┬───┘    └──┬───┘
    │           │           │
    └───────────┼───────────┘
                ▼
         ┌─────────────┐
         │  Combiner   │
         │ (synthesize │
         │  final ans) │
         └─────────────┘
```

**Key Steps:**
1. Identify natural decomposition points
2. Solve each sub-problem independently
3. Combine results into final answer
4. Optionally verify the combined solution

## 💡 Basic Example

Demonstrating decomposed prompting on a multi-part analysis question.

**Problem:** Compare two smartphones across price, camera, battery, and performance.

````python
# Decomposed Prompting Exampledef solve_subproblem(subproblem, context=""):    """Solve a single subproblem"""    prompt = f"""{context}Sub-problem: {subproblem}Provide a concise answer:"""    return get_completion(prompt, temperature=0.3)def combine_solutions(sub_solutions, original_problem):    """Combine sub-problem solutions into final answer"""    combined = "".join([f"{k}:{v}" for k, v in sub_solutions.items()])    prompt = f"""Original problem: {original_problem}Here are the solutions to each sub-problem:{combined}Synthesize these into a comprehensive final answer with clear recommendations:"""    return get_completion(prompt, temperature=0.3)# Main problemmain_problem = "Compare iPhone 15 and Samsung Galaxy S24 for a professional photographer."print("="*60)print("🔨 Decomposed Prompting")print("="*60)print(f"Main Problem: {main_problem}")# Define sub-problemssub_problems = {    "Camera Quality": "Compare the camera specifications and photo quality of iPhone 15 vs Samsung Galaxy S24.",    "Storage Options": "Compare storage capacities and pricing for both phones.",    "Battery Life": "Compare battery life and charging capabilities.",    "Ecosystem Integration": "Compare how well each phone integrates with professional photography workflows."}# Solve each sub-problemprint("Solving sub-problems...")solutions = {}for name, subproblem in sub_problems.items():    print(f"--- {name} ---")    solution = solve_subproblem(subproblem, f"Context: {main_problem}")    solutions[name] = solution    print(solution[:200] + "..." if len(solution) > 200 else solution + "")# Combine solutionsprint("="*60)print("📊 Final Combined Answer:")print("="*60)final_answer = combine_solutions(solutions, main_problem)print(final_answer)
````

## ⚠️ Failure Case

Decomposed Prompting can fail when:
1. Sub-problems are not truly independent
2. The decomposition loses important context
3. Combining solutions introduces contradictions
4. Overhead exceeds benefits for simple problems
5. Sub-problem solutions are inconsistent

````python
# Failure Case: When decomposition hurts# Problem where decomposition loses contextcontext_dependent_problem = """Sarah is taller than Mike. Mike is taller than John. Is Sarah taller than John? Explain why."""print("="*60)print("⚠️ Decomposition Failure Case")print("="*60)print(f"Problem: {context_dependent_problem.strip()}")# Show simple solutionprint("✅ Simple approach (works perfectly):")simple = get_completion(context_dependent_problem, temperature=0)print(simple)# Show failed decomposed approachprint("❌ Decomposed approach (unnecessary and potentially confusing):")sub1 = get_completion("Is Sarah taller than Mike?", temperature=0)sub2 = get_completion("Is Mike taller than John?", temperature=0)print(f"Sub-problem 1: {sub1}")print(f"Sub-problem 2: {sub2}")combine = get_completion(f"""Based on:- {sub1}- {sub2}Is Sarah taller than John?""", temperature=0)print(f"Combined: {combine}")print("💡 Lesson: Don't decompose when:")print("   - Problem is already simple")print("   - Sub-problems lose transitive/contextual relationships")print("   - Single reasoning chain is sufficient")
````

## 💡 Tips & Tricks

**Best Practices:**
- ✓ Identify truly independent sub-problems
- ✓ Pass relevant context to each sub-problem
- ✓ Design clear combination/synthesis step
- ✓ Use consistent formatting across sub-problems
- ✓ Consider parallel execution for efficiency
- ✗ Don't decompose simple problems
- ✗ Don't create too many sub-problems (>5 is usually too many)
- ✗ Don't lose interdependencies between sub-problems

**When to Use:**
- Multi-part questions
- Analysis with distinct dimensions
- Problems requiring different expertise areas
