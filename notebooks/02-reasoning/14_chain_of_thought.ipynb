# Chain-of-Thought (CoT)

**Category**: Reasoning & Logic  
**Technique #**: 14

## 📖 Description
Encouraging the model to show its step-by-step reasoning process before arriving at a final answer.

**When to use**: Math word problems, logical reasoning, complex decision-making, any task requiring multi-step logic.

## ⚙️ How It Works
1. **Present the problem** — Give the model a question or task.
2. **Trigger reasoning** — Add 'Let\'s think step by step' or provide CoT examples.
3. **Model reasons aloud** — The model generates intermediate reasoning steps.
4. **Final answer** — The model arrives at the answer after its reasoning chain.

```
┌──────────┐    ┌──────────────┐    ┌──────────┐    ┌────────┐
│ Question  │───▶│  Step 1...   │───▶│  Step N  │───▶│ Answer │
└──────────┘    │  Step 2...   │    └──────────┘    └────────┘
                └──────────────┘
```

## 🔧 Setup
Install required libraries and set up API keys.

````python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
````

## 🧪 Basic Example
Solving a math problem with step-by-step reasoning.

````python
# Chain-of-Thought example
prompt = """Solve this problem step by step:

A store sells apples for $2 each. If you buy 5 or more, you get a 20% discount.
How much would 7 apples cost?

Let's think step by step:"""

response = ask(prompt, temperature=0)
print(response)
````

## 🌍 Real‑World Example
Using CoT for complex business decision analysis.

````python
# Real-world: Business decision analysis with CoT
prompt = """A company is deciding whether to launch a new product.

Data:
- Development cost: $500,000
- Expected unit price: $49.99
- Manufacturing cost per unit: $15
- Marketing budget: $200,000
- Market research shows potential demand: 30,000 units in year 1
- Competitor offers similar product at $59.99

Should the company launch this product? Think step by step,
considering profitability, market positioning, and risks."""

response = ask(prompt, temperature=0.3)
print(response)
````

## ❌ Failure Case
Can produce plausible-sounding but incorrect reasoning chains. Adds verbosity to simple factual queries.

````python
# Failure case for Chain-of-Thought (CoT)
# Why it fails: Can produce plausible-sounding but incorrect reasoning chains. Adds verbosity to simple factual queries.

failure_prompt = """Using the Chain-of-Thought (CoT) approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Can produce plausible-sounding but incorrect reasoning chains. Adds verbosity to...")
````

## 📊 Benchmark
GSM8K: CoT improves GPT-4 from ~70% to ~92%. BBH: +10-15% across diverse reasoning tasks.

````python
# Benchmark comparison for Chain-of-Thought (CoT)
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Chain-of-Thought (CoT)
    enhanced = ask(f"Using Chain-of-Thought (CoT), {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: GSM8K: CoT improves GPT-4 from ~70% to ~92%. BBH: +10-15% across diverse reasoni")
````

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Adding 'Let\'s think step by step' alone boosts reasoning. For best results, provide CoT examples.

**Claude**: Use `<thinking>` tags to encourage structured reasoning before the final answer.

**Gemini**: Explicitly ask for numbered steps. Gemini responds well to structured reasoning requests.

**Parameter Tips**: Use `temperature=0` for math/logic, `temperature=0.3` for analytical tasks.

**Common Pitfalls**: CoT can produce confident but wrong reasoning. Always verify the final answer independently.
