## 📋 Description

Program-Aided LLM (PAL) offloads calculations and algorithmic reasoning to a programming runtime instead of relying on the LLM for computation. The LLM generates code (Python, SQL, etc.) which is then executed to produce accurate results.

**When to use:**
- Mathematical calculations and word problems
- Data analysis and manipulation
- Algorithmic reasoning
- Problems requiring precise computation
- Any task where exact answers matter

## 🔧 How It Works

```
PAL Process:

┌─────────────────────────────────────────────────────┐
│  Problem: "If a train travels 120 miles in 2        │
│           hours, what's its speed?"                │
└──────────────────┬──────────────────────────────────┘
                   │
                   ▼
         ┌─────────────────┐
         │  LLM generates  │
         │  Python code    │
         └────────┬────────┘
                  │
    distance = 120
    time = 2
    speed = distance / time
    print(speed)
                  │
                  ▼
         ┌─────────────────┐
         │  Python Runtime │
         │  executes code  │
         └────────┬────────┘
                  │
                  ▼
              Output: 60.0
```

**Key Benefits:**
- Eliminates arithmetic errors from LLMs
- Enables complex data processing
- Provides reproducible results
- Handles large numbers precisely

## 💡 Basic Example

Comparing direct LLM reasoning vs PAL for a math word problem.

**Problem:** A store sells 150 items per day. Each item costs $12 to make and sells for $25. What is the daily profit?

````python
# Program-Aided LLM (PAL) Example# Problemproblem = """A store sells 150 items per day. Each item costs $12 to make and sells for $25. What is the daily profit?"""print("="*60)print("🔢 Program-Aided LLM (PAL) Demo")print("="*60)# Method 1: Direct LLM reasoningprint("❌ Method 1: Direct LLM Reasoning")print("-" * 40)direct_prompt = f"""{problem}Calculate the answer step by step."""direct_answer = get_completion(direct_prompt, temperature=0)print(direct_answer)# Method 2: PAL - Generate Python codeprint("✅ Method 2: PAL (Generate & Execute Python)")print("-" * 40)pal_prompt = f"""{problem}Write Python code to solve this problem. Store the final answer in a variable called 'result' and print it.Python code:"""generated_code = get_completion(pal_prompt, temperature=0)print("Generated Code:")print("```python")print(generated_code)print("```")# Execute the generated code safelyprint("Executing code...")try:    # Create a safe execution environment    exec_globals = {}    exec(generated_code, exec_globals)    # Try to get the result    if 'result' in exec_globals:        print(f"✅ Answer: ${exec_globals['result']}")    else:        print("✅ Code executed successfully")except Exception as e:    print(f"❌ Error executing code: {e}")# Method 3: Using a structured PAL approachprint("" + "="*60)print("🎯 Structured PAL Approach")print("="*60)def pal_solve(problem):    """Structured PAL solver"""    # Step 1: Generate code    code_prompt = f"""Problem: {problem}Generate Python code to solve this. Follow these rules:1. Define all variables clearly2. Show intermediate calculations3. Store final answer in variable 'answer'4. Print the answer with explanationPython code:"""    code = get_completion(code_prompt, temperature=0)    # Step 2: Execute code    print("Generated Code:")    print(code)    print("Execution Result:")    try:        exec_globals = {}        exec(code, exec_globals)        return exec_globals.get('answer', 'See output above')    except Exception as e:        return f"Error: {e}"result = pal_solve(problem)print(f"Final Answer: {result}")
````

## ⚠️ Failure Case

1. The LLM generates incorrect or buggy code
2. Required libraries aren't available
3. The code has syntax errors
4. The problem is better suited for natural language reasoning
5. Security concerns prevent code execution

## 💡 Tips & Tricks

**Implementation Tips:**
- ✓ Use clear variable names in prompts
- ✓ Request intermediate calculations be shown
- ✓ Handle errors gracefully in execution
- ✓ Validate generated code before execution
- ✓ Use restricted execution environments for safety
- ✗ Don't use for non-computational problems
- ✗ Don't execute untrusted code without sandboxing

**Supported Languages:**
- Python (most common)
- SQL (for database queries)
- JavaScript (for web-related tasks)
- Bash (for system operations)
