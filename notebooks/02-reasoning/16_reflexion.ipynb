## 📋 Description

Reflexion is a self-reflective approach where language models learn from their mistakes by generating verbal reinforcement signals. After attempting a task and receiving feedback, the model reflects on what went wrong and updates its strategy for future attempts.

**When to use:**
- Iterative improvement tasks
- Complex problems requiring multiple attempts
- When you can provide binary or graded feedback
- Learning from demonstration scenarios
- Tasks with clear success/failure criteria

## 🔧 How It Works

```
Reflexion Loop:

┌─────────────────────────────────────────────────────┐
│              Initial Attempt                        │
│  Task: Write a Python function to sort a list       │
│  Output: def sort_list(lst):                        │
│              return lst.sort()  ← Bug!              │
└──────────────────┬──────────────────────────────────┘
                   │
                   ▼
         ┌─────────────────┐
         │    Evaluate     │
         │  (Test & Score) │
         └────────┬────────┘
                  │
         Result: FAILED
         Error: Returns None
                  │
                  ▼
         ┌─────────────────┐
         │    Reflect      │
         │ (What went      │
         │  wrong?)        │
         └────────┬────────┘
                  │
         "lst.sort() sorts in-place
          and returns None. Should
          use sorted(lst) instead."
                  │
                  ▼
         ┌─────────────────┐
         │  Second Attempt │
         │ (with lessons   │
         │  learned)       │
         └────────┬────────┘
                  │
         def sort_list(lst):
             return sorted(lst)  ✓
                  │
                  ▼
              SUCCESS!
```

**Key Components:**
1. Actor - generates the initial attempt
2. Evaluator - assesses the output
3. Self-Reflection - analyzes failures
4. Memory - stores lessons learned

Implementing Reflexion for code generation with iterative improvement.

**Task:** Write a function to calculate the factorial of a number.

````python
# Reflexion: Self-Reflective Improvementclass ReflexionAgent:    def __init__(self):        self.reflection_memory = []    def generate_attempt(self, task, previous_reflections=[]):        """Generate an attempt at the task"""        reflection_text = ""        if previous_reflections:            reflection_text = "Previous reflections:" + "".join(previous_reflections)        prompt = f"""Task: {task}{reflection_text}Provide your solution:"""        return get_completion(prompt, temperature=0.3)    def evaluate(self, task, attempt):        """Evaluate the attempt (simplified - in practice, run tests)"""        prompt = f"""Task: {task}Attempted Solution:{attempt}Evaluate this solution. Does it correctly solve the task?If not, what errors or issues do you see?Evaluation:"""        return get_completion(prompt, temperature=0.3)    def reflect(self, task, attempt, evaluation):        """Generate reflection on what went wrong"""        prompt = f"""Task: {task}Attempted Solution:{attempt}Evaluation:{evaluation}Reflect on what went wrong and how to improve. What specific lessons can be learned?Reflection:"""        return get_completion(prompt, temperature=0.3)    def solve(self, task, max_attempts=3):        """Main Reflexion loop"""        print("="*60)        print("🔄 Reflexion Agent")        print("="*60)        print(f"Task: {task}")        for attempt_num in range(1, max_attempts + 1):            print(f"--- Attempt {attempt_num} ---")            # Generate attempt            attempt = self.generate_attempt(task, self.reflection_memory)            print(f"📝 Solution:{attempt}")            # Evaluate            evaluation = self.evaluate(task, attempt)            print(f"📊 Evaluation:{evaluation}")            # Check if successful            if "correct" in evaluation.lower() and "error" not in evaluation.lower():                print("✅ Task completed successfully!")                return attempt            # Reflect and store lesson            if attempt_num < max_attempts:                reflection = self.reflect(task, attempt, evaluation)                print(f"🤔 Reflection:{reflection}")                self.reflection_memory.append(reflection)        print("⚠️ Max attempts reached")        return attempt# Test Reflexionagent = ReflexionAgent()task = """Write a Python function called 'factorial' that:- Takes a non-negative integer n as input- Returns n! (n factorial)- Returns 1 for n = 0- Raises ValueError for negative inputs"""final_solution = agent.solve(task, max_attempts=3)
````

## 🌍 Real-World Example

**Scenario:** Email writing with iterative refinement based on feedback.

Reflexion helps improve the tone, clarity, and professionalism of business communications.

````python
# Real-World: Email Writing with Reflexionprint("="*60)print("📧 Email Refinement with Reflexion")print("="*60)email_task = """Write a professional email to a client (Acme Corp) informing them that their project delivery will be delayed by 2 weeks due to unforeseen technical challenges.Maintain a positive relationship while being transparent about the delay."""# Attempt 1print("📝 Attempt 1:")print("-" * 40)attempt1 = get_completion(f"""{email_task}Write the email:""", temperature=0.7)print(attempt1)# Evaluationprint("📊 Evaluation:")print("-" * 40)evaluation1 = get_completion(f"""Review this email for:1. Professional tone2. Clarity about the delay3. Empathy for the client4. Proposed solution/next stepsEmail:{attempt1}Evaluation:""", temperature=0.3)print(evaluation1)# Reflectionprint("🤔 Reflection:")print("-" * 40)reflection1 = get_completion(f"""Based on this evaluation:{evaluation1}What specific improvements should be made to the email?Reflection:""", temperature=0.3)print(reflection1)# Attempt 2 with reflectionprint("📝 Attempt 2 (incorporating reflection):")print("-" * 40)attempt2 = get_completion(f"""{email_task}Previous reflection on how to improve:{reflection1}Write an improved email:""", temperature=0.7)print(attempt2)# Final evaluationprint("📊 Final Evaluation:")print("-" * 40)final_eval = get_completion(f"""Compare these two emails for the same task:EMAIL 1:{attempt1}EMAIL 2:{attempt2}Which is better and why? What improvements were made?""", temperature=0.3)print(final_eval)
````

## ⚠️ Failure Case

Reflexion can fail when:
1. The evaluation criteria are unclear
2. Reflections don't capture the actual problem
3. The model repeats the same mistakes despite reflection
4. There's no clear success metric
5. The task is too complex for iterative refinement

````python
# Failure Case: Reflexion limitationsprint("="*60)print("⚠️ Reflexion Failure Cases")print("="*60)# Case 1: Unclear evaluationprint("❌ Case 1: Unclear success criteria")print("-" * 40)unclear_task = """Write a creative story about a robot."""print(f"Task: {unclear_task.strip()}")print("💡 Problem: What makes a story 'good'? Without clear criteria,")print("   reflections may not lead to meaningful improvements.")# Case 2: Same mistake repeatedprint("❌ Case 2: Inability to learn from reflection")print("-" * 40)repeated_mistake = """Example where reflection doesn't help:Attempt 1: Code with off-by-one errorReflection: "I need to check array bounds"Attempt 2: Code with same off-by-one errorReflection: "I need to be more careful with indices"Attempt 3: Code STILL has off-by-one error💡 Problem: The model 'knows' the issue but doesn't apply the fix!"""print(repeated_mistake)print("" + "="*60)print("💡 When Reflexion Works Best:")print("="*60)print("✓ Clear success criteria (tests, rubrics)")print("✓ Specific, actionable feedback")print("✓ Model can actually implement the fix")print("✓ Iterative improvement is possible")print("✗ Not suitable for:")print("  - Open-ended creative tasks")print("  - Problems requiring external knowledge")print("  - Tasks where model lacks capability")
````

## 💡 Tips & Tricks

**Implementation Tips:**
- ✓ Provide clear evaluation criteria
- ✓ Use automated testing when possible
- ✓ Store and reference previous reflections
- ✓ Limit iterations to prevent loops
- ✓ Make reflections specific and actionable
- ✗ Don't use for purely creative tasks
- ✗ Don't expect improvement if model lacks capability

**Best Practices:**
1. Start with binary feedback (success/failure)
2. Progress to graded feedback (1-10 scale)
3. Include examples of good reflections
4. Consider using separate models for actor/evaluator/reflector
