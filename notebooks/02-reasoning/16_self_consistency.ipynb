## 📖 Description
Generating multiple reasoning paths and selecting the most common answer via majority voting.

**When to use**: High-stakes reasoning tasks where a single chain of thought may be unreliable.

## ⚙️ How It Works
**Step-by-step process for Self-Consistency (CoT-SC):**

1. **Prepare** — Understand the task requirements and constraints.
2. **Construct the prompt** — Apply the Self-Consistency (CoT-SC) technique to structure your prompt.
3. **Send to model** — Submit the crafted prompt to the LLM.
4. **Evaluate output** — Assess whether the response meets expectations.
5. **Iterate** — Refine the prompt based on results if needed.

## 🔧 Setup
Install required libraries and set up API keys.

```python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
```

## 🧪 Basic Example
Generating multiple reasoning paths and finding the consensus answer.

```python
# Self-consistency: generate multiple answers and vote
prompt = """Solve step by step:
If a train travels 120 miles in 2 hours, then stops for 30 minutes,
then travels 90 miles in 1.5 hours, what is the average speed
for the entire journey (including the stop)?

Think step by step:"""

# Generate 5 responses
answers = []
for i in range(5):
    resp = ask(prompt, temperature=0.7)
    answers.append(resp)
    print(f"--- Response {i+1} ---")
    print(resp)
    print()

print("\n=== Majority vote: review the answers above to find consensus ===")
```

## 🌍 Real‑World Example
Applying Self-Consistency (CoT-SC) to a practical high-stakes reasoning tasks where a single chain of thought may be unreliable. scenario.

```python
# Real-world Self-Consistency (CoT-SC) application
# Use case: High-stakes reasoning tasks where a single chain of thought may be unreliable.

prompt = """You are helping with a real-world task.
Apply the Self-Consistency (CoT-SC) technique.

Task: Analyze the following customer support ticket and provide:
1. Issue category
2. Priority level (High/Medium/Low)
3. Suggested resolution
4. Estimated time to resolve

Ticket: 'My account has been locked for 3 days and I cannot access my
subscription content. I have a presentation tomorrow and need my files.
I already tried resetting my password twice.'

Provide your analysis:"""

response = ask(prompt, temperature=0.3)
print(response)
```

## ❌ Failure Case
Computationally expensive (requires multiple API calls). Less effective when all paths share the same error.

```python
# Failure case for Self-Consistency (CoT-SC)
# Why it fails: Computationally expensive (requires multiple API calls). Less effective when all paths share the same error.

failure_prompt = """Using the Self-Consistency (CoT-SC) approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Computationally expensive (requires multiple API calls). Less effective when all...")
```

## 📊 Benchmark
GSM8K: Self-consistency with 40 paths achieves ~96% with GPT-4, +4% over single CoT.

```python
# Benchmark comparison for Self-Consistency (CoT-SC)
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Self-Consistency (CoT-SC)
    enhanced = ask(f"Using Self-Consistency (CoT-SC), {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: GSM8K: Self-consistency with 40 paths achieves ~96% with GPT-4, +4% over single ")
```

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Best results with clear formatting. Experiment with system messages to set up Self-Consistency (CoT-SC).

**Claude**: Claude tends to follow instructions precisely. Use XML tags for structure when applying Self-Consistency (CoT-SC).

**Gemini**: Works well with explicit step-by-step instructions for Self-Consistency (CoT-SC).

**Parameter Tuning**:
- `temperature=0`: Deterministic, best for factual tasks
- `temperature=0.3-0.5`: Balanced creativity and consistency
- `temperature=0.7-1.0`: More creative, diverse outputs

**Common Pitfalls**: Over-engineering the prompt can backfire. Start simple and iterate.
