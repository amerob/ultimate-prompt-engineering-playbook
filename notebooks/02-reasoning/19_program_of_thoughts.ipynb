## 📖 Description
Having the model generate executable code (usually Python) to solve reasoning problems instead of natural language chains.

**When to use**: Mathematical calculations, data processing, algorithmic problems where computation is needed.

## ⚙️ How It Works
**Step-by-step process for Program of Thoughts (PoT):**

1. **Prepare** — Understand the task requirements and constraints.
2. **Construct the prompt** — Apply the Program of Thoughts (PoT) technique to structure your prompt.
3. **Send to model** — Submit the crafted prompt to the LLM.
4. **Evaluate output** — Assess whether the response meets expectations.
5. **Iterate** — Refine the prompt based on results if needed.

## 🔧 Setup
Install required libraries and set up API keys.

```python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
```

## 🧪 Basic Example
A minimal example demonstrating the Program of Thoughts (PoT) technique.

```python
# Basic Program of Thoughts (PoT) example
prompt = """Apply the Program of Thoughts (PoT) technique to answer the following:

What are the three main causes of climate change?

Provide a clear, structured answer."""

response = ask(prompt, temperature=0.3)
print(response)
```

```python
# Failure case for Program of Thoughts (PoT)
# Why it fails: Requires a code execution environment. May generate buggy code. Not suitable for subjective reasoning.

failure_prompt = """Using the Program of Thoughts (PoT) approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Requires a code execution environment. May generate buggy code. Not suitable for...")
```

## 📊 Benchmark
GSM8K: PoT achieves ~96% with GPT-4 by offloading computation to Python interpreter.

```python
# Benchmark comparison for Program of Thoughts (PoT)
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Program of Thoughts (PoT)
    enhanced = ask(f"Using Program of Thoughts (PoT), {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: GSM8K: PoT achieves ~96% with GPT-4 by offloading computation to Python interpre")
```

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Best results with clear formatting. Experiment with system messages to set up Program of Thoughts (PoT).

**Claude**: Claude tends to follow instructions precisely. Use XML tags for structure when applying Program of Thoughts (PoT).

**Gemini**: Works well with explicit step-by-step instructions for Program of Thoughts (PoT).

**Parameter Tuning**:
- `temperature=0`: Deterministic, best for factual tasks
- `temperature=0.3-0.5`: Balanced creativity and consistency
- `temperature=0.7-1.0`: More creative, diverse outputs

**Common Pitfalls**: Over-engineering the prompt can backfire. Start simple and iterate.
