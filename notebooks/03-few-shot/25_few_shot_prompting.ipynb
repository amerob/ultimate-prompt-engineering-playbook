**Category**: Few-Shot & Example-Based  
**Technique #**: 25

## 📖 Description
Providing a small number of input-output examples before the actual task to guide the model's behavior.

**When to use**: Classification, formatting, translation, any task where demonstrating the pattern is faster than describing it.

## ⚙️ How It Works
1. **Select examples** — Choose 2-5 representative input-output pairs.
2. **Format consistently** — Present examples in a clear, consistent structure.
3. **Add the actual query** — Append the real task after the examples.
4. **Model follows pattern** — The model mimics the demonstrated pattern.

```
┌─────────────┐    ┌─────────────┐    ┌─────────┐    ┌────────┐
│ Example 1   │    │ Example 2   │    │ Query   │───▶│ Answer │
│ Input→Output│    │ Input→Output│    │         │    │        │
└─────────────┘    └─────────────┘    └─────────┘    └────────┘
```

## 🔧 Setup
Install required libraries and set up API keys.

````python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
````

## 🧪 Basic Example
Demonstrating the pattern with a few examples before the actual task.

````python
# Few-shot sentiment classification
prompt = """Classify the sentiment of the text.

Text: "I love this product!" → Sentiment: POSITIVE
Text: "Terrible experience, never again." → Sentiment: NEGATIVE
Text: "It was okay, nothing special." → Sentiment: NEUTRAL

Text: "This exceeded all my expectations!" → Sentiment:"""

response = ask(prompt, temperature=0)
print(response)
````

## 🌍 Real‑World Example
Applying Few-Shot Prompting to a practical classification scenario.

````python
# Real-world Few-Shot Prompting application
# Use case: Classification

prompt = """You are helping with a real-world task.
Apply the Few-Shot Prompting technique.

Task: Analyze the following customer support ticket and provide:
1. Issue category
2. Priority level (High/Medium/Low)
3. Suggested resolution
4. Estimated time to resolve

Ticket: 'My account has been locked for 3 days and I cannot access my
subscription content. I have a presentation tomorrow and need my files.
I already tried resetting my password twice.'

Provide your analysis:"""

response = ask(prompt, temperature=0.3)
print(response)
````

## ❌ Failure Case
Poor example selection can mislead the model. Examples may bias toward specific patterns.

````python
# Failure case for Few-Shot Prompting
# Why it fails: Poor example selection can mislead the model. Examples may bias toward specific patterns.

failure_prompt = """Using the Few-Shot Prompting approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Poor example selection can mislead the model. Examples may bias toward specific ...")
````

## 📊 Benchmark
SuperGLUE: Few-shot GPT-4 approaches fine-tuned model performance on many benchmarks.

````python
# Benchmark comparison for Few-Shot Prompting
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Few-Shot Prompting
    enhanced = ask(f"Using Few-Shot Prompting, {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: SuperGLUE: Few-shot GPT-4 approaches fine-tuned model performance on many benchm")
````

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Best results with clear formatting. Experiment with system messages to set up Few-Shot Prompting.

**Claude**: Claude tends to follow instructions precisely. Use XML tags for structure when applying Few-Shot Prompting.

**Gemini**: Works well with explicit step-by-step instructions for Few-Shot Prompting.

**Parameter Tuning**:
- `temperature=0`: Deterministic, best for factual tasks
- `temperature=0.3-0.5`: Balanced creativity and consistency
- `temperature=0.7-1.0`: More creative, diverse outputs

**Common Pitfalls**: Over-engineering the prompt can backfire. Start simple and iterate.
