In-Context Instruction Learning combines natural language instructions with examples, teaching models both what to do and how to do it through paired instruction-example demonstrations.

## Description

In-Context Instruction Learning pairs each example with an instruction:

- **Instruction + Example**: Each demonstration includes both
- **Meta-learning**: Model learns the task structure
- **Better generalization**: Instructions help transfer learning
- **Improved accuracy**: Combines benefits of both approaches

**When to Use:**
- Complex multi-step tasks
- Need for explicit reasoning
- Teaching new task types
- Improving few-shot robustness

## How It Works

```
IN-CONTEXT INSTRUCTION LEARNING STRUCTURE

Standard Few-Shot:
  Input: Hello
  Output: Bonjour
  
  Input: Goodbye
  Output: Au revoir

Instruction + Example:
  Instruction: Translate from English to French
  Input: Hello
  Output: Bonjour
  
  Instruction: Translate from English to French
  Input: Goodbye
  Output: Au revoir

Benefits:
- Clear task definition
- Better generalization
- Explicit reasoning path
- Improved accuracy
```

## Setup

````python
!pip install openai -q

import os
from getpass import getpass
from openai import OpenAI

api_key = getpass("Enter your OpenAI API key: ")
os.environ["OPENAI_API_KEY"] = api_key

client = OpenAI()

def get_completion(prompt, model="gpt-4"):
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
````

## Basic Example: With vs Without Instructions

Compare standard few-shot with instruction-based few-shot.

````python
# Standard few-shot (without explicit instructions per example)
standard_prompt = """
Classify the sentiment:

Text: I love this product!
Sentiment: Positive

Text: This is terrible.
Sentiment: Negative

Text: It's okay, I guess.
Sentiment:
"""

# Instruction-based few-shot
instruction_prompt = """
Task: Classify text sentiment as Positive or Negative

Example 1:
Instruction: Determine if the sentiment is positive or negative
Text: I love this product!
Sentiment: Positive

Example 2:
Instruction: Determine if the sentiment is positive or negative
Text: This is terrible.
Sentiment: Negative

Task: Classify text sentiment as Positive or Negative
Text: It's okay, I guess.
Sentiment:
"""

print("=== STANDARD FEW-SHOT ===")
print(get_completion(standard_prompt))

print("\n=== INSTRUCTION-BASED FEW-SHOT ===")
print(get_completion(instruction_prompt))
````

## Real-World Example: Multi-Step Reasoning

Using instruction learning for complex reasoning tasks.

````python
# Multi-step reasoning with instruction learning
reasoning_prompt = """
Task: Solve math word problems step by step

Example 1:
Instruction: Break down the problem and solve step by step
Problem: John has 5 apples. He buys 3 more and gives away 2. How many does he have?
Solution:
Step 1: Start with 5 apples
Step 2: Add 3 apples: 5 + 3 = 8
Step 3: Subtract 2 apples: 8 - 2 = 6
Answer: 6 apples

Example 2:
Instruction: Break down the problem and solve step by step
Problem: A train travels 60 km/h for 2 hours, then 80 km/h for 3 hours. Total distance?
Solution:
Step 1: First segment: 60 km/h * 2 h = 120 km
Step 2: Second segment: 80 km/h * 3 h = 240 km
Step 3: Total: 120 + 240 = 360 km
Answer: 360 km

Task: Solve math word problems step by step
Problem: Sarah earns $15/hour. She works 8 hours on Monday and 6 hours on Tuesday. Total earnings?
Solution:
"""

print("=== INSTRUCTION-BASED REASONING ===")
print(get_completion(reasoning_prompt))
````

## Failure Case: Conflicting Instructions

When instructions in examples conflict with each other or the task.

````python
# Demonstrate conflicting instructions
conflicting_prompt = """
Task: Summarize the text in one sentence

Example 1:
Instruction: Provide a detailed summary with all key points
Text: The quick brown fox jumps over the lazy dog.
Summary: A quick brown fox jumps over a lazy dog.

Example 2:
Instruction: Summarize in exactly 5 words
Text: The cat sat on the mat.
Summary: Cat sat on the mat

Task: Summarize the text in one sentence
Text: Machine learning is transforming how we build software.
Summary:
"""

print("=== CONFLICTING INSTRUCTIONS ===")
print(get_completion(conflicting_prompt))

print("\n⚠️ Problem: Conflicting instructions cause inconsistent output!")
print("\n✅ Solution: Keep instructions consistent across all examples")
````

## Interactive Playground

Build your own instruction-based few-shot prompt.

````python
# Interactive instruction learning builder
task = input("Define your task: ")
instruction = input("Instruction to include with each example: ")

num_examples = int(input("\nNumber of examples: "))

examples = []
for i in range(num_examples):
    print(f"\n--- Example {i+1} ---")
    inp = input("Input: ")
    out = input("Output: ")
    examples.append((inp, out))

test_input = input("\nTest input: ")

# Build instruction-based prompt
prompt = f"Task: {task}\n\n"
for i, (inp, out) in enumerate(examples):
    prompt += f"Example {i+1}:\n"
    prompt += f"Instruction: {instruction}\n"
    prompt += f"Input: {inp}\n"
    prompt += f"Output: {out}\n\n"
prompt += f"Task: {task}\n"
prompt += f"Input: {test_input}\n"
prompt += "Output:"

print("\n=== GENERATED PROMPT ===")
print(prompt)
print("\n=== MODEL OUTPUT ===")
print(get_completion(prompt))
````

## Tips & Tricks

### Best Practices

1. **Consistent instructions**: Same instruction format for all examples
2. **Clear task definition**: Explicitly state what to do
3. **Reasoning steps**: Include thinking process for complex tasks
4. **Format consistency**: Same structure across examples

### Instruction Templates

**Classification**: "Classify the following as [categories]"

**Generation**: "Generate a [output type] based on the input"

**Transformation**: "Transform the input to [target format]"

**Reasoning**: "Solve step by step, showing your work"

### Model-Specific Notes

**GPT-4**: Excellent at following instruction patterns

**Claude**: Benefits from detailed, explicit instructions

**GPT-3.5**: Needs very clear, consistent instruction formats
