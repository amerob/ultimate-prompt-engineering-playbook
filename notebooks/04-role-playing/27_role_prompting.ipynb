## Description

Role Prompting is a fundamental technique where you explicitly assign a specific role, profession, or identity to the AI. By telling the model to "act as" a particular persona, you activate relevant knowledge patterns, adjust the tone and style of responses, and guide the model to approach problems from a specific perspective.

### When to Use:
- Need specialized expertise or domain knowledge
- Want responses in a specific professional style
- Require perspective from a particular viewpoint
- Need creative content from a specific character
- Want to constrain responses to a professional framework

### Key Benefits:
✓ Activates domain-specific knowledge
✓ Sets appropriate tone and vocabulary
✓ Provides implicit constraints and frameworks
✓ Improves response relevance and depth

## How It Works

```
┌─────────────────────────────────────────────────────────────┐
│                     ROLE PROMPTING FLOW                     │
└─────────────────────────────────────────────────────────────┘

  ┌──────────────┐         ┌──────────────┐         ┌─────────────┐
  │   User       │────────▶│   Assign     │────────▶│   Model     │
  │   Request    │         │   Role       │         │   Activates │
  └──────────────┘         └──────────────┘         └─────────────┘
                                                            │
                                                            ▼
  ┌──────────────┐         ┌──────────────┐         ┌─────────────┐
  │   Receive    │◀────────│   Generate   │◀────────│   Role      │
  │   Response   │         │   Output     │         │   Knowledge │
  └──────────────┘         └──────────────┘         └─────────────┘
```

### The Mechanics:
1. **Role Assignment**: User specifies a role ("Act as a...")
2. **Knowledge Activation**: Model retrieves relevant training patterns
3. **Framework Application**: Model adopts appropriate methodology
4. **Tone Adjustment**: Language style matches the role
5. **Response Generation**: Output reflects the assigned persona

## Basic Example

Simple role assignment for a financial advisor explaining investment concepts.

````python
def get_completion(prompt, role=None, model="gpt-4o-mini"):
    """Get completion with optional role assignment."""
    messages = []
    
    if role:
        # Add system message with role
        messages.append({
            "role": "system",
            "content": f"You are a {role}. Respond accordingly."
        })
    
    messages.append({"role": "user", "content": prompt})
    
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0.7
    )
    
    return response.choices[0].message.content

# Test without role
print("═" * 60)
print("WITHOUT ROLE (Generic Response):")
print("═" * 60)
prompt = "Explain compound interest to a beginner."
response_no_role = get_completion(prompt)
print(response_no_role)

print("\n" + "═" * 60)
print("WITH ROLE (Financial Advisor):")
print("═" * 60)
response_with_role = get_completion(prompt, role="seasoned financial advisor")
print(response_with_role)
````

## Real-World Example: Technical Documentation

Using role prompting to generate documentation from different professional perspectives.

````python
# Technical topic to explain
topic = "REST API authentication using JWT tokens"

# Different roles for different audiences
roles = {
    "Senior Software Architect": "For engineering team leads",
    "Technical Writer": "For developer documentation",
    "DevOps Engineer": "For infrastructure/security teams",
    "Junior Developer Mentor": "For new programmers"
}

print(f"Topic: {topic}\n")

for role, audience in roles.items():
    print(f"\n{'='*70}")
    print(f"ROLE: {role} ({audience})")
    print(f"{'='*70}")
    
    prompt = f"Explain {topic} in 3-4 sentences."
    response = get_completion(prompt, role=role)
    print(response[:300] + "..." if len(response) > 300 else response)
````

## Failure Case: When Role Prompting Falls Short

Role prompting has limitations. Here are scenarios where it may fail:

````python
# Failure Case 1: Contradictory roles
print("FAILURE CASE 1: CONTRADICTORY ROLES")
print("═" * 60)

bad_role = "honest politician who always lies"
prompt = "Tell me about tax policy."
response = get_completion(prompt, role=bad_role)
print(f"Role: {bad_role}")
print(f"Response: {response[:200]}...")
print("\n⚠️  Issue: Contradictory instructions confuse the model\n")

# Failure Case 2: Overly vague role
print("FAILURE CASE 2: VAGUE ROLE")
print("═" * 60)

vague_role = "expert"
prompt = "Analyze this data."
response = get_completion(prompt, role=vague_role)
print(f"Role: {vague_role}")
print(f"Response: {response[:200]}...")
print("\n⚠️  Issue: 'Expert' is too generic - no specific domain activated")
````

## Tips & Tricks

### Best Practices:

1. **Be Specific**: "Senior Python Developer" > "Developer"
2. **Add Qualifiers**: Include experience level, specialty, or style
3. **Use "Act as" Format**: "Act as a..." is most effective
4. **Combine with Context**: Role + specific instructions work best
5. **Test Multiple Roles**: Compare outputs from different personas

```

• You are a [specialty] expert specializing in [area]

```
