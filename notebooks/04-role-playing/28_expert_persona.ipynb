## Description

Expert Persona takes role prompting to a deeper level by creating a detailed, multi-dimensional expert character. Instead of simply assigning a job title, you define the expert's background, specializations, communication style, and approach to problems. This creates more authentic, nuanced responses that reflect genuine expertise.

### When to Use:
- Need highly specialized, technical responses
- Want authoritative analysis with professional depth
- Require industry-specific terminology and frameworks
- Need nuanced understanding of complex topics
- Want to simulate consultation with a real expert

### Key Benefits:
✓ Activates deeper domain knowledge
✓ Provides professional methodologies and frameworks
✓ Creates more authoritative and credible responses
✓ Enables sophisticated problem-solving approaches

4. **Establish Methodology**: How they approach problems
5. **Add Personality Nuance**: Professional philosophy, values

## Basic Example

Creating a detailed cybersecurity expert persona for security analysis.

````python
def create_expert_persona(expert_definition, user_prompt, model="gpt-4o-mini"):
    """
    Create a response using a detailed expert persona.
    
    Args:
        expert_definition: Detailed description of the expert
        user_prompt: The user's question or request
        model: OpenAI model to use
    """
    messages = [
        {
            "role": "system",
            "content": expert_definition
        },
        {
            "role": "user",
            "content": user_prompt
        }
    ]
    
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0.7
    )
    
    return response.choices[0].message.content

# Define a detailed cybersecurity expert
cybersecurity_expert = """
You are Dr. Sarah Chen, a cybersecurity expert with the following profile:

CREDENTIALS:
- PhD in Computer Science from MIT, specializing in network security
- 15 years of experience in enterprise security architecture
- Former Chief Security Officer at Fortune 500 tech company
- CISSP, CEH, and OSCP certified

SPECIALIZATIONS:
- Zero-trust architecture implementation
- Threat intelligence and incident response
- Cloud security (AWS, Azure, GCP)
- Penetration testing and vulnerability assessment

COMMUNICATION STYLE:
- Professional yet accessible to technical audiences
- Uses industry-standard terminology (MITRE ATT&CK, CVSS, etc.)
- Provides actionable recommendations with risk assessments
- References real-world examples and case studies

APPROACH:
- Defense-in-depth methodology
- Risk-based prioritization
- Balances security with usability
"""

# Test the expert persona
prompt = "A company wants to migrate to cloud infrastructure. What are the top 5 security considerations?"

print("═" * 70)
print("EXPERT PERSONA RESPONSE:")
print("═" * 70)
response = create_expert_persona(cybersecurity_expert, prompt)
print(response)
````

## Real-World Example: Multi-Domain Analysis

Using different expert personas to analyze the same business problem from multiple professional perspectives.

````python
# Define multiple expert personas for a business scenario
experts = {
    "CFO": """
    You are Michael Rodriguez, CFO with 20 years experience in financial leadership.
    MBA from Wharton, former CFO at two successful startups (one IPO).
    Specializations: Financial modeling, capital allocation, investor relations.
    Style: Data-driven, focuses on ROI, cash flow, and financial metrics.
    Approach: Conservative financial planning with strategic growth investments.
    """,
    
    "CTO": """
    You are Dr. Priya Sharma, CTO with 18 years in technology leadership.
    PhD in AI/ML from Stanford, led engineering at Google and Meta.
    Specializations: AI/ML architecture, scalable systems, tech strategy.
    Style: Technical depth balanced with business acumen.
    Approach: Innovation-focused, emphasizes technical debt and scalability.
    """,
    
    "CMO": """
    You are Jennifer Park, CMO with 15 years in marketing leadership.
    Built brands at Nike and Airbnb, led marketing for 3 unicorn startups.
    Specializations: Brand strategy, growth marketing, customer acquisition.
    Style: Creative and customer-centric, uses marketing frameworks.
    Approach: Brand-first thinking with data-driven optimization.
    """
}

# Business scenario to analyze
business_scenario = """
Our SaaS company ($10M ARR, 200 employees) is considering:
1. Expanding into European markets (estimated $2M investment)
2. Building an AI feature set (estimated $3M investment)
3. Doubling down on US market share (estimated $2M investment)

We have $5M available. Which strategy would you recommend and why?
"""

print("BUSINESS SCENARIO ANALYSIS")
print("═" * 70)
print(business_scenario)
print("═" * 70)

# Get analysis from each expert
for title, persona in experts.items():
    print(f"\n{'='*70}")
    print(f"{title} PERSPECTIVE")
    print(f"{'='*70}")
    
    response = create_expert_persona(persona, business_scenario)
    print(response[:500] + "...\n" if len(response) > 500 else response + "\n")
````

## Failure Case: Expert Persona Pitfalls

Understanding when expert personas fail and why.

````python
# Failure Case 1: Impossibly broad expertise
print("FAILURE CASE 1: IMPOSSIBLY BROAD EXPERTISE")
print("═" * 70)

bad_expert = """
You are Dr. Universal Expert, a world-renowned authority in:
- Quantum physics, neurosurgery, ancient history, cryptocurrency,
  marine biology, constitutional law, and medieval poetry.
"""

prompt = "Explain the connection between quantum entanglement and medieval poetry."
response = create_expert_persona(bad_expert, prompt)
print(response[:300])
print("\n⚠️  Issue: No single person has such diverse expertise - response lacks depth\n")

# Failure Case 2: Overly restrictive constraints
print("FAILURE CASE 2: OVERLY RESTRICTIVE CONSTRAINTS")
print("═" * 70)

restrictive_expert = """
You are a medical doctor who:
- Only speaks in 5-word sentences
- Never uses medical terminology
- Must include a joke in every response
- Cannot mention any body parts
"""

prompt = "What causes headaches?"
response = create_expert_persona(restrictive_expert, prompt)
print(response[:300])
print("\n⚠️  Issue: Too many constraints prevent effective communication")
````

## Tips & Tricks

### Building Effective Expert Personas:

1. **Specific Credentials Matter**: "Harvard MBA with 15 years at McKinsey" > "Business expert"
2. **Define 2-3 Specializations**: Prevents overly broad responses
3. **Include Communication Style**: Formal, conversational, technical, accessible
4. **Add Professional Philosophy**: How they approach problems
5. **Reference Real Frameworks**: Encourages use of established methodologies

### Expert Persona Template:

```
You are [Name], a [profession] with:

BACKGROUND:
- [Education/Training]
- [Years of experience]
- [Notable positions/companies]

EXPERTISE:
- [Specialization 1]
- [Specialization 2]
- [Specialization 3]

STYLE:
- [Communication approach]
- [Tone characteristics]

METHOD:
- [Professional approach/framework]
```

### Common Pitfalls to Avoid:
- ❌ Impossibly broad expertise
- ❌ Contradictory characteristics
- ❌ Overly restrictive constraints
- ❌ Vague credentials
- ❌ Too many specializations (dilutes focus)
