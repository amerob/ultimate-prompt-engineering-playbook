## 📖 Description
Assigning the model a specific role or identity (e.g., 'You are a senior data scientist') to shape responses.

**When to use**: Domain expertise simulation, customer service personas, educational tutoring.

## ⚙️ How It Works
**Step-by-step process for Persona Prompting:**

1. **Prepare** — Understand the task requirements and constraints.
2. **Construct the prompt** — Apply the Persona Prompting technique to structure your prompt.
3. **Send to model** — Submit the crafted prompt to the LLM.
4. **Evaluate output** — Assess whether the response meets expectations.
5. **Iterate** — Refine the prompt based on results if needed.

## 🔧 Setup
Install required libraries and set up API keys.

```python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
```

## 🧪 Basic Example
Assigning an expert persona to improve domain-specific responses.

```python
# Persona prompting example
system_msg = """You are a senior Python developer with 15 years of experience.
You write clean, well-documented code following PEP 8 standards.
You always consider edge cases and error handling."""

prompt = "Write a function to validate an email address."

response = ask(prompt, system=system_msg, temperature=0.3)
print(response)
```

## 🌍 Real‑World Example
Applying Persona Prompting to a practical domain expertise simulation scenario.

```python
# Real-world Persona Prompting application
# Use case: Domain expertise simulation

prompt = """You are helping with a real-world task.
Apply the Persona Prompting technique.

Task: Analyze the following customer support ticket and provide:
1. Issue category
2. Priority level (High/Medium/Low)
3. Suggested resolution
4. Estimated time to resolve

Ticket: 'My account has been locked for 3 days and I cannot access my
subscription content. I have a presentation tomorrow and need my files.
I already tried resetting my password twice.'

Provide your analysis:"""

response = ask(prompt, temperature=0.3)
print(response)
```

## ❌ Failure Case
Model may over-commit to persona and refuse valid requests. Persona knowledge limited by training data.

```python
# Failure case for Persona Prompting
# Why it fails: Model may over-commit to persona and refuse valid requests. Persona knowledge limited by training data.

failure_prompt = """Using the Persona Prompting approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Model may over-commit to persona and refuse valid requests. Persona knowledge li...")
```

## 📊 Benchmark
Qualitative: improves domain-specific response quality by ~20% in human evaluation.

```python
# Benchmark comparison for Persona Prompting
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Persona Prompting
    enhanced = ask(f"Using Persona Prompting, {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: Qualitative: improves domain-specific response quality by ~20% in human evaluati")
```

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Best results with clear formatting. Experiment with system messages to set up Persona Prompting.

**Claude**: Claude tends to follow instructions precisely. Use XML tags for structure when applying Persona Prompting.

**Gemini**: Works well with explicit step-by-step instructions for Persona Prompting.

**Parameter Tuning**:
- `temperature=0`: Deterministic, best for factual tasks
- `temperature=0.3-0.5`: Balanced creativity and consistency
- `temperature=0.7-1.0`: More creative, diverse outputs

**Common Pitfalls**: Over-engineering the prompt can backfire. Start simple and iterate.
