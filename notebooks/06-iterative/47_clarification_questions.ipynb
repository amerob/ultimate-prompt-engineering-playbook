## 📋 Description

**Clarification Questions** is a technique where the AI asks targeted questions to gather missing information before providing a response. This transforms the interaction from a one-shot query into a collaborative dialogue that produces more accurate and relevant outputs.

### When to Use:
- Ambiguous user requests
- Complex problem-solving tasks
- Requirements gathering
- Content creation with specific constraints
- Any scenario where context is incomplete

## 🎯 Basic Example

Simple example showing AI asking clarification questions.

````
def ask_with_clarification(user_request, max_questions=3):
    """
    Generate a response that first asks clarifying questions.
    
    Args:
        user_request: The initial user query
        max_questions: Maximum number of questions to ask
    """
    
    clarification_prompt = f"""
    The user has made the following request: "{user_request}"
    
    Before answering, identify what information is missing or ambiguous.
    Ask up to {max_questions} targeted clarification questions that would
    help you provide the most accurate and helpful response.
    
    Format your response as:
    QUESTIONS:
    1. [First question]
    2. [Second question]
    etc.
    """
    
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[{"role": "user", "content": clarification_prompt}]
    )
    
    return response.choices[0].message.content

# Example: Vague request
vague_request = "Write me a marketing email."

print("=" * 60)
print("VAGUE REQUEST:")
print("=" * 60)
print(f"User: {vague_request}\n")

questions = ask_with_clarification(vague_request)
print("AI CLARIFICATION QUESTIONS:")
print(questions)

# Now simulate user providing answers
print("\n" + "=" * 60)
print("SIMULATED USER ANSWERS:")
print("=" * 60)

answers = """
1. It's for a software company launching a new project management tool.
2. The target audience is small business owners and freelancers.
3. The tone should be professional but friendly, highlighting time-saving benefits.
"""

print(answers)

# Generate final response with clarifications
final_prompt = f"""
Write a marketing email based on this request: {vague_request}

Use these clarifications:
{answers}
"""

final_response = client.chat.completions.create(
    model="gpt-4o",
    messages=[{"role": "user", "content": final_prompt}]
)

print("\n" + "=" * 60)
print("FINAL TAILORED RESPONSE:")
print("=" * 60)
print(final_response.choices[0].message.content)
````

## 💼 Real-World Example

Requirements gathering for a software project.

````
# Software requirements gathering
project_request = "I need a website for my business."

print("=" * 60)
print("PROJECT REQUIREMENTS GATHERING")
print("=" * 60 + "\n")

# Step 1: Ask comprehensive questions
requirements_prompt = f"""
You are a senior business analyst helping a client define their website requirements.

Client request: "{project_request}"

Ask 5-7 specific questions covering:
- Business type and goals
- Target audience
- Required features and functionality
- Design preferences
- Budget and timeline
- Technical requirements

Make your questions specific and actionable.
"""

questions_response = client.chat.completions.create(
    model="gpt-4o",
    messages=[{"role": "user", "content": requirements_prompt}]
)

print("BUSINESS ANALYST QUESTIONS:")
print(questions_response.choices[0].message.content)

# Step 2: Simulate client answers
print("\n" + "=" * 60)
print("CLIENT ANSWERS:")
print("=" * 60)

client_answers = """
1. I run a boutique fitness studio offering yoga, pilates, and personal training.
2. Main goal is to increase class bookings and attract new members.
3. Target audience is health-conscious professionals aged 25-45, mostly women.
4. Need: class schedule, online booking, instructor profiles, pricing, contact form.
5. Prefer clean, modern design with calming colors (greens, whites).
6. Budget is $5,000-8,000, timeline is 6-8 weeks.
7. Should be mobile-friendly and integrate with Mindbody for bookings.
"""

print(client_answers)

# Step 3: Generate comprehensive requirements document
print("\n" + "=" * 60)
print("GENERATED REQUIREMENTS DOCUMENT:")
print("=" * 60 + "\n")

doc_prompt = f"""
Based on the following client information, create a professional
website requirements document.

Original request: {project_request}

Client answers:
{client_answers}

Include:
1. Executive Summary
2. Business Objectives
3. Target Audience Analysis
4. Functional Requirements
5. Design Requirements
6. Technical Specifications
7. Timeline and Budget
"""

doc_response = client.chat.completions.create(
    model="gpt-4o",
    messages=[{"role": "user", "content": doc_prompt}]
)

print(doc_response.choices[0].message.content[:1500] + "...")
````

## ⚠️ Failure Case

When clarification questions go wrong and how to fix them.

````
# ❌ BAD: Too many questions, overwhelming the user
bad_clarification = """
❌ BAD PRACTICE - Asking Too Many Questions:

User: "Help me plan a trip"

AI: I need more information:
1. Where do you want to go?
2. When are you traveling?
3. How long is your trip?
4. What's your budget?
5. Who is traveling with you?
6. What type of accommodation do you prefer?
7. Do you need flights?
8. What activities interest you?
9. Do you have dietary restrictions?
10. What's your preferred travel style?
11. Do you need travel insurance?
12. What languages do you speak?
... (15+ more questions)

PROBLEM: Overwhelming, may discourage the user
"""

print(bad_clarification)

# ✅ GOOD: Prioritized, limited questions
good_clarification = """
✅ GOOD PRACTICE - Focused, Prioritized Questions:

User: "Help me plan a trip"

AI: I'd love to help! To give you the best recommendations,
    let me start with 3 quick questions:

1. What's your destination (or type of destination)?
2. When and for how long are you traveling?
3. What's your approximate budget range?

We can discuss activities, accommodation, and other details next!

BENEFIT: Gets essential info without overwhelming
"""

print(good_clarification)

print("\n" + "=" * 60)
print("BEST PRACTICES FOR CLARIFICATION QUESTIONS:")
print("=" * 60)
print("""

✅ DO:
   - Limit to 2-4 questions per round
   - Prioritize most critical information first
   - Make questions specific and actionable
   - Group related questions together
   - Explain why you need the information

❌ DON'T:
   - Ask more than 5 questions at once
   - Ask questions that could be answered with assumptions
   - Use jargon or technical terms
   - Ask redundant or overlapping questions
   - Make users feel interrogated

""")

# Demonstrate the good approach
print("\n" + "=" * 60)
print("DEMONSTRATION - GOOD APPROACH:")
print("=" * 60 + "\n")

good_prompt = """
The user said: "Help me plan a trip"

Ask up to 3 focused clarification questions that will help you
provide the most useful initial recommendations. Prioritize the
most important factors for trip planning.
"""

good_response = client.chat.completions.create(
    model="gpt-4o",
    messages=[{"role": "user", "content": good_prompt}]
)

print(good_response.choices[0].message.content)
````

## 💡 Tips & Tricks

### Best Practices:

1. **Prioritize ruthlessly** - Ask only what's essential
2. **Batch strategically** - Group related questions
3. **Explain value** - Tell users why you need the info
4. **Offer defaults** - Suggest common options

### Question Types by Scenario:

| Scenario | Key Questions |
|----------|---------------|
| Content Creation | Audience, tone, length, purpose |
| Technical Help | Environment, error messages, goal |
| Planning | Timeline, budget, constraints |
| Advice | Context, preferences, constraints |

### Template for System Prompts:
```
Before answering any request:
1. Identify what's missing or ambiguous
2. Ask up to 3 focused questions
3. Wait for answers before proceeding
```
