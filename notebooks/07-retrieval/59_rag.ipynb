## 📖 Description
Retrieving relevant documents from an external knowledge base and including them in the prompt context.

**When to use**: Q&A over private documents, up-to-date information retrieval, enterprise knowledge bases.

## ⚙️ How It Works
1. **User query** — Receive the user's question.
2. **Retrieve** — Search a knowledge base for relevant documents.
3. **Augment** — Insert retrieved documents into the prompt context.
4. **Generate** — Model answers grounded in the provided documents.

```
┌───────┐    ┌────────────┐    ┌──────────────┐    ┌──────────┐
│ Query │───▶│  Retriever  │───▶│ Prompt +     │───▶│ Answer   │
│       │    │  (vector DB)│    │ Documents    │    │(grounded)│
└───────┘    └────────────┘    └──────────────┘    └──────────┘
```

## 🔧 Setup
Install required libraries and set up API keys.

````python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
````

## 🧪 Basic Example
Augmenting a prompt with retrieved context to ground the model's response.

````python
# Simulated RAG example
# In production, these documents would come from a vector database
retrieved_docs = [
    "Document 1: The Eiffel Tower was built in 1889 for the World's Fair in Paris.",
    "Document 2: The Eiffel Tower is 330 meters tall and was the tallest structure until 1930.",
    "Document 3: Gustave Eiffel's company designed and built the tower."
]

context = '\n'.join(retrieved_docs)
prompt = f"""Based ONLY on the following documents, answer the question.
If the answer is not in the documents, say 'I don't have enough information.'

Documents:
{context}

Question: Who designed the Eiffel Tower and when was it built?"""

response = ask(prompt, temperature=0)
print(response)
````

## 🌍 Real‑World Example
Applying Retrieval-Augmented Generation (RAG) to a practical q&a over private documents scenario.

````python
# Real-world Retrieval-Augmented Generation (RAG) application
# Use case: Q&A over private documents

prompt = """You are helping with a real-world task.
Apply the Retrieval-Augmented Generation (RAG) technique.

Task: Analyze the following customer support ticket and provide:
1. Issue category
2. Priority level (High/Medium/Low)
3. Suggested resolution
4. Estimated time to resolve

Ticket: 'My account has been locked for 3 days and I cannot access my
subscription content. I have a presentation tomorrow and need my files.
I already tried resetting my password twice.'

Provide your analysis:"""

response = ask(prompt, temperature=0.3)
print(response)
````

## ❌ Failure Case
Retrieved documents may be irrelevant. Context window limits the amount of retrieved text.

````python
# Failure case for Retrieval-Augmented Generation (RAG)
# Why it fails: Retrieved documents may be irrelevant. Context window limits the amount of retrieved text.

failure_prompt = """Using the Retrieval-Augmented Generation (RAG) approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Retrieved documents may be irrelevant. Context window limits the amount of retri...")
````

## 📊 Benchmark
NaturalQuestions: RAG improves accuracy by ~20-30% over closed-book generation.

````python
# Benchmark comparison for Retrieval-Augmented Generation (RAG)
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Retrieval-Augmented Generation (RAG)
    enhanced = ask(f"Using Retrieval-Augmented Generation (RAG), {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: NaturalQuestions: RAG improves accuracy by ~20-30% over closed-book generation.")
````

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Best results with clear formatting. Experiment with system messages to set up Retrieval-Augmented Generation (RAG).

**Claude**: Claude tends to follow instructions precisely. Use XML tags for structure when applying Retrieval-Augmented Generation (RAG).

**Gemini**: Works well with explicit step-by-step instructions for Retrieval-Augmented Generation (RAG).

**Parameter Tuning**:
- `temperature=0`: Deterministic, best for factual tasks
- `temperature=0.3-0.5`: Balanced creativity and consistency
- `temperature=0.7-1.0`: More creative, diverse outputs

**Common Pitfalls**: Over-engineering the prompt can backfire. Start simple and iterate.
