## 📋 Description

Self-Correction is a prompting technique where the model is asked to review and correct its own output. Instead of accepting the first response, the model evaluates its answer for errors, inconsistencies, or improvements and produces a refined version.

**When to Use:**
- Complex reasoning tasks where errors are likely
- Mathematical or logical problem-solving
- Content that requires factual accuracy
- Code generation and debugging
- When high precision is critical

## 🔧 How It Works

```
┌─────────────────┐     ┌─────────────────┐     ┌─────────────────┐
│  Initial Query  │────▶│  First Response  │────▶│ Self-Evaluation │
└─────────────────┘     └─────────────────┘     └─────────────────┘
                                                        │
                                                        ▼
                                              ┌─────────────────┐
                                              │  Identify Issues │
                                              └─────────────────┘
                                                        │
                                                        ▼
                                              ┌─────────────────┐
                                              │ Corrected Output │
                                              └─────────────────┘
```

**The Process:**
1. Generate initial response to the prompt
2. Ask the model to review its own output critically
3. Identify errors, omissions, or improvements
4. Produce a corrected and refined final answer

## 💡 Basic Example

````python
def self_correction_example():
    """Demonstrate basic self-correction technique."""
    
    # Step 1: Initial response
    initial_prompt = """
    Calculate: 247 × 38
    Show your work step by step.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": initial_prompt}]
    )
    
    initial_answer = response1.choices[0].message.content
    print("📤 INITIAL RESPONSE:")
    print(initial_answer)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Self-correction
    correction_prompt = f"""
    You previously gave this answer:
    """
    {initial_answer}
    """
    
    Please review your work carefully. Check for:
    - Calculation errors
    - Arithmetic mistakes
    - Logic errors
    
    If you find any errors, provide the corrected answer with explanation.
    If no errors, confirm the answer is correct.
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": correction_prompt}]
    )
    
    corrected_answer = response2.choices[0].message.content
    print("🔍 SELF-CORRECTION:")
    print(corrected_answer)
    
    return initial_answer, corrected_answer

# Run the example
initial, corrected = self_correction_example()
````

## 🌍 Real-World Example: Business Email

````python
def business_email_with_self_correction():
    """Real-world example: Writing and correcting a business email."""
    
    # Step 1: Draft email
    draft_prompt = """
    Write a professional email to a client informing them that their
    project delivery will be delayed by 2 weeks due to supply chain issues.
    The original deadline was March 15th.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": draft_prompt}]
    )
    
    draft_email = response1.choices[0].message.content
    print("📧 INITIAL EMAIL DRAFT:")
    print(draft_email)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Self-correction for tone and professionalism
    correction_prompt = f"""
    Review this email draft you wrote:
    """
    {draft_email}
    """
    
    Critically evaluate it for:
    1. Tone - Is it appropriately professional yet empathetic?
    2. Clarity - Is the delay reason clearly explained?
    3. Solutions - Are alternatives or mitigation strategies offered?
    4. Accountability - Is responsibility appropriately taken?
    5. New deadline - Is a revised timeline provided?
    
    Provide an improved version addressing any issues you identify.
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": correction_prompt}]
    )
    
    corrected_email = response2.choices[0].message.content
    print("✅ CORRECTED EMAIL:")
    print(corrected_email)
    
    return draft_email, corrected_email

# Run the example
draft, corrected = business_email_with_self_correction()
````

## ⚠️ Failure Case: When Self-Correction Fails

````python
def failure_case_example():
    """Demonstrate when self-correction may not work."""
    
    # Problem: The model may confidently maintain incorrect answers
    prompt = """
    What is the capital of Australia?
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}]
    )
    
    answer1 = response1.choices[0].message.content
    print(f"Initial answer: {answer1}")
    
    # Attempt correction with misleading hint
    correction = """
    You said: "{answer1}"
    
    Wait, I think the capital might actually be Melbourne.
    Can you verify and correct if needed?
    """.format(answer1=answer1)
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": correction}]
    )
    
    answer2 = response2.choices[0].message.content
    print(f"After 'correction': {answer2}")
    
    print("\n⚠️ Note: Models can be swayed by incorrect 'corrections' from users,")
    print("or may fail to catch subtle errors in their reasoning.")

failure_case_example()
````

## 💡 Tips & Tricks

### Model-Specific Advice

**GPT-4/GPT-4o:**
- Excellent at self-correction
- Can identify subtle logical errors
- Works well with explicit error-checking instructions

**Claude:**
- Strong self-reflection capabilities
- Good at catching factual inconsistencies
- Responds well to structured correction prompts

**Gemini:**
- Effective for mathematical self-correction
- Good at code review and debugging
- May need more explicit correction instructions

### Best Practices
1. **Be specific** about what to check for
2. **Use structured prompts** with clear evaluation criteria
3. **Multiple rounds** may be needed for complex problems
4. **Temperature settings** - use lower temp (0.1-0.3) for more consistent corrections
5. **Combine with other techniques** like Chain-of-Thought for best results
