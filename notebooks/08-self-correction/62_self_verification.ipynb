## 📋 Description

Self-Verification is a technique where the model checks its own answers for correctness before finalizing them. Unlike self-correction which produces a new answer, verification focuses on validating the existing answer and providing a confidence assessment.

**When to Use:**
- Before finalizing critical answers
- When factual accuracy is paramount
- In multi-step reasoning tasks
- For verification of computational results
- When you need a confidence score

## 🔧 How It Works

```
┌─────────────────┐     ┌─────────────────┐     ┌─────────────────┐
│  Generate Answer │────▶│  Self-Verify    │────▶│  Pass/Fail?     │
└─────────────────┘     └─────────────────┘     └─────────────────┘
                              │                          │
                              ▼                          ▼
                        ┌─────────┐                ┌─────────┐
                        │ Check   │                │ Output  │
                        │ Facts   │────────PASS───▶│ Result  │
                        │ Logic   │                │ + Conf. │
                        │ Consistency              └─────────┘
                        └─────────┘                      │
                              │                          │
                              └────FAIL────▶ [Retry/Flag]
```

**The Process:**
1. Generate an answer to the query
2. Systematically verify each component
3. Check facts, logic, and consistency
4. Provide verification result with confidence score

## 💡 Basic Example

````python
def self_verification_example():
    """Demonstrate self-verification with confidence scoring."""
    
    # Combined generation and verification
    prompt = """
    Answer this question, then verify your answer:
    
    Question: What is the square root of 144 plus the cube root of 27?
    
    Format your response as:
    ANSWER: [Your answer]
    VERIFICATION: [Step-by-step verification]
    CONFIDENCE: [High/Medium/Low with percentage]
    """
    
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}]
    )
    
    result = response.choices[0].message.content
    print("📊 GENERATION + VERIFICATION:")
    print(result)
    
    return result

# Run the example
self_verification_example()
````

## 🌍 Real-World Example: Medical Information Verification

````python
def medical_info_verification():
    """Real-world example: Verifying health information."""
    
    # Step 1: Generate information
    query = """
    What are the common side effects of ibuprofen?
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": query}]
    )
    
    info = response1.choices[0].message.content
    print("📋 INITIAL INFORMATION:")
    print(info)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Verification with structured criteria
    verification_prompt = f"""
    Verify the following medical information for accuracy:
    """
    {info}
    """
    
    Check against these criteria:
    1. FACTUAL ACCURACY: Are the stated side effects correct?
    2. COMPLETENESS: Are major side effects missing?
    3. SEVERITY INDICATION: Is severity properly indicated?
    4. FREQUENCY: Are common vs. rare side effects distinguished?
    5. CONTRAINDICATIONS: Are important warnings mentioned?
    
    Provide:
    - VERIFICATION STATUS: [VERIFIED / PARTIALLY VERIFIED / NEEDS REVIEW]
    - CONFIDENCE SCORE: [0-100%]
    - ISSUES FOUND: [List any concerns]
    - RECOMMENDATIONS: [Suggested improvements]
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": verification_prompt}]
    )
    
    verification = response2.choices[0].message.content
    print("🔍 VERIFICATION RESULT:")
    print(verification)
    
    return info, verification

# Run the example
info, verification = medical_info_verification()
````

## ⚠️ Failure Case: Overconfidence in Verification

````python
def verification_failure_case():
    """Show when self-verification can fail."""
    
    # The model may verify confidently even when wrong
    prompt = """
    Answer and verify: What is 2 + 2 × 2?
    
    First give your answer, then verify it step by step.
    """
    
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}]
    )
    
    result = response.choices[0].message.content
    print("Result:")
    print(result)
    
    print("\n⚠️ Note: Even with verification, models can:")
    print("   - Be overconfident in incorrect answers")
    print("   - Verify using the same flawed reasoning")
    print("   - Miss subtle errors in their logic")
    print("\n   (Correct answer: 2 + (2 × 2) = 6, following order of operations)")

verification_failure_case()
````

## 💡 Tips & Tricks

### Model-Specific Advice

**GPT-4/GPT-4o:**
- Excellent at structured verification
- Good at catching logical inconsistencies
- Use system prompts to enforce verification behavior

**Claude:**
- Strong at nuanced verification
- Good at expressing uncertainty appropriately
- Effective for complex reasoning verification

**Gemini:**
- Good for factual verification
- Effective for code verification
- May need explicit confidence scoring instructions

### Best Practices
1. **Always ask for confidence scores** - forces model to assess certainty
2. **Use structured verification criteria** - be specific about what to check
3. **Separate generation and verification** - reduces confirmation bias
4. **Multiple verification rounds** - for critical information
5. **Cross-reference when possible** - compare against known facts
