## 📋 Description

Critique & Revise is a structured self-improvement technique where the model first generates output, then critiques it from multiple perspectives, and finally revises based on the critique. This mimics the human writing process of draft → review → revision.

**When to Use:**
- Creative writing and content creation
- Argument development and essay writing
- Business proposals and presentations
- Code review and optimization
- Any task where quality can be improved through iteration

## 🔧 How It Works

```
┌─────────────┐    ┌─────────────┐    ┌─────────────┐    ┌─────────────┐
│   DRAFT    │───▶│  CRITIQUE  │───▶│   REVISE   │───▶│   FINAL    │
│  (Create)  │    │  (Analyze) │    │ (Improve)  │    │  (Polish)  │
└─────────────┘    └─────────────┘    └─────────────┘    └─────────────┘
                         │
                         ▼
                   Multiple Perspectives:
                   - Strengths
                   - Weaknesses
                   - Missing elements
                   - Improvement suggestions
```

**The Process:**
1. **Draft**: Generate initial output
2. **Critique**: Analyze from multiple angles (strengths, weaknesses, gaps)
3. **Revise**: Produce improved version addressing critiques
4. **Final**: Polish and finalize the output

## 💡 Basic Example

````python
def critique_and_revise_example():
    """Demonstrate the critique and revise technique."""
    
    # Step 1: Generate initial draft
    draft_prompt = """
    Write a short paragraph (3-4 sentences) about the importance of
    learning to code in the modern world.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": draft_prompt}]
    )
    
    draft = response1.choices[0].message.content
    print("📝 DRAFT:")
    print(draft)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Critique the draft
    critique_prompt = f"""
    Critique this paragraph about learning to code:
    """
    {draft}
    """
    
    Provide a detailed critique covering:
    1. STRENGTHS: What works well?
    2. WEAKNESSES: What could be improved?
    3. MISSING ELEMENTS: What's absent that would strengthen it?
    4. CLARITY: Is the message clear and compelling?
    5. SPECIFIC SUGGESTIONS: Concrete improvements to make
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": critique_prompt}]
    )
    
    critique = response2.choices[0].message.content
    print("🔍 CRITIQUE:")
    print(critique)
    print("\n" + "="*50 + "\n")
    
    # Step 3: Revise based on critique
    revise_prompt = f"""
    Original draft:
    """
    {draft}
    """
    
    Critique received:
    """
    {critique}
    """
    
    Now write a revised version (3-4 sentences) that addresses
    the critique while maintaining the core message.
    """
    
    response3 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": revise_prompt}]
    )
    
    revised = response3.choices[0].message.content
    print("✅ REVISED VERSION:")
    print(revised)
    
    return draft, critique, revised

# Run the example
draft, critique, revised = critique_and_revise_example()
````

## 🌍 Real-World Example: Marketing Copy

````python
def marketing_copy_critique():
    """Real-world: Creating marketing copy with critique and revision."""
    
    # Step 1: Draft marketing copy
    draft_prompt = """
    Write marketing copy for a new productivity app called 'FocusFlow'
    that helps people manage their time better. Keep it under 100 words.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": draft_prompt}]
    )
    
    draft = response1.choices[0].message.content
    print("📝 INITIAL MARKETING COPY:")
    print(draft)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Multi-perspective critique
    critique_prompt = f"""
    Critique this marketing copy from multiple perspectives:
    """
    {draft}
    """
    
    PERSPECTIVE 1 - MARKETER: Does it sell? Is the value prop clear?
    PERSPECTIVE 2 - CUSTOMER: Would this convince me to try it?
    PERSPECTIVE 3 - EDITOR: Is it well-written and persuasive?
    PERSPECTIVE 4 - COMPETITOR: What would competitors say?
    
    For each perspective, note strengths and specific improvements needed.
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": critique_prompt}]
    )
    
    critique = response2.choices[0].message.content
    print("🔍 MULTI-PERSPECTIVE CRITIQUE:")
    print(critique)
    print("\n" + "="*50 + "\n")
    
    # Step 3: Final revision
    final_prompt = f"""
    Original marketing copy:
    """
    {draft}
    """
    
    Critique received:
    """
    {critique}
    """
    
    Write the FINAL VERSION (under 100 words) that:
    - Addresses all major critique points
    - Maintains a compelling, persuasive tone
    - Has a clear call-to-action
    - Is polished and professional
    """
    
    response3 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": final_prompt}]
    )
    
    final = response3.choices[0].message.content
    print("✅ FINAL MARKETING COPY:")
    print(final)
    
    return draft, critique, final

# Run the example
draft, critique, final = marketing_copy_critique()
````

## ⚠️ Failure Case: Over-Critiquing

````python
def critique_failure_case():
    """Show when critique can be counterproductive."""
    
    # Sometimes excessive critique leads to losing the original voice
    prompt = """
    Write a short, punchy tagline for a coffee brand.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}]
    )
    
    draft = response1.choices[0].message.content
    print(f"Draft: {draft}")
    
    # Overly harsh critique
    critique_prompt = f"""
    Critique this tagline harshly: "{draft}"
    
    List everything wrong with it and why it would fail.
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": critique_prompt}]
    )
    
    critique = response2.choices[0].message.content
    print(f"\nCritique: {critique}")
    
    # Revision may lose originality
    revise_prompt = f"""
    Original: {draft}
    Critique: {critique}
    
    Rewrite completely based on the critique.
    """
    
    response3 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": revise_prompt}]
    )
    
    revised = response3.choices[0].message.content
    print(f"\nRevised: {revised}")
    
    print("\n⚠️ Note: Overly negative critique can lead to:")
    print("   - Loss of original voice/creativity")
    print("   - Generic, safe outputs")
    print("   - Over-engineered solutions")

critique_failure_case()
````

## 🎮 Interactive Playground

````python
# 🎮 Interactive Playground

YOUR_WRITING_TASK = """
[Describe what you want to write - e.g., "A product description for..."]
"""

def custom_critique_revise(task, perspectives=None):
    """Apply critique and revise to any writing task."""
    
    if perspectives is None:
        perspectives = ["General Quality", "Clarity", "Impact"]
    
    # Step 1: Draft
    draft_prompt = f"""
    {task}
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": draft_prompt}]
    )
    draft = response1.choices[0].message.content
    print("📝 DRAFT:")
    print(draft)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Critique
    critique_prompt = f"""
    Critique this content:
    """
    {draft}
    """
    
    Critique from these perspectives:
    """
    for i, p in enumerate(perspectives, 1):
        critique_prompt += f"{i}. {p}\n"
    
    critique_prompt += """
    
    For each: note strengths, weaknesses, and specific improvements.
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": critique_prompt}]
    )
    critique = response2.choices[0].message.content
    print("🔍 CRITIQUE:")
    print(critique)
    print("\n" + "="*50 + "\n")
    
    # Step 3: Revise
    revise_prompt = f"""
    Original:
    """
    {draft}
    """
    
    Critique:
    """
    {critique}
    """
    
    Write a revised version addressing the critique.
    """
    
    response3 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": revise_prompt}]
    )
    revised = response3.choices[0].message.content
    print("✅ REVISED:")
    print(revised)
    
    return draft, critique, revised

# Uncomment to run
# custom_critique_revise(YOUR_WRITING_TASK)
````

## 💡 Tips & Tricks

### Model-Specific Advice

**GPT-4/GPT-4o:**
- Excellent at providing balanced critiques
- Good at maintaining voice during revision
- Use temperature 0.7 for creative tasks, 0.3 for technical

**Claude:**
- Strong at nuanced, constructive criticism
- Good at preserving original intent
- Effective for sensitive content review

**Gemini:**
- Good for structured critique frameworks
- Effective for technical content revision
- May need explicit balance instructions

### Best Practices
1. **Balance critique** - ask for strengths AND weaknesses
2. **Specific perspectives** - define what angles to critique from
3. **Preserve voice** - explicitly ask to maintain original style
4. **Multiple rounds** - 2-3 iterations often optimal
5. **Know when to stop** - over-iteration can degrade quality
