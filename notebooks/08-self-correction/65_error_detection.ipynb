## 📋 Description

Error Detection is a technique where the model systematically reviews its output to identify mistakes, inconsistencies, or problems. Unlike general self-correction, this focuses specifically on finding errors before they reach the user.

**When to Use:**
- Code generation and debugging
- Mathematical calculations
- Fact-checking and verification
- Logical reasoning tasks
- Multi-step problem solving

## 🔧 How It Works

```
┌─────────────────┐     ┌─────────────────┐     ┌─────────────────┐
│  Generate Output│────▶│  Error Detection │────▶│  Errors Found?  │
└─────────────────┘     └─────────────────┘     └─────────────────┘
                              │                          │
                              ▼                          ▼
                        ┌──────────┐              ┌──────────┐
                        │ Check:   │              │ List     │
                        │ - Facts  │              │ Errors   │
                        │ - Logic  │────YES─────▶│ & Fixes  │
                        │ - Math   │              │          │
                        │ - Syntax │              │          │
                        └──────────┘              └──────────┘
                              │
                              └────NO────▶ [Output Verified]
```

**The Process:**
1. Generate initial output
2. Systematically scan for errors:
   - Factual errors
   - Logical inconsistencies
   - Calculation mistakes
   - Syntax/grammar issues
3. Report findings with corrections

## 💡 Basic Example

````python
def error_detection_example():
    """Demonstrate error detection in calculations."""
    
    # Step 1: Generate a calculation with potential errors
    problem = """
    Calculate: (125 + 75) × 4 - 200 ÷ 5
    Show each step.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": problem}]
    )
    
    solution = response1.choices[0].message.content
    print("📋 INITIAL SOLUTION:")
    print(solution)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Error detection
    detection_prompt = f"""
    Review this solution for errors:
    """
    {solution}
    """
    
    Check for:
    1. CALCULATION ERRORS - Any arithmetic mistakes?
    2. ORDER OF OPERATIONS - Followed correctly?
    3. LOGICAL ERRORS - Any reasoning problems?
    4. TRANSCRIPTION ERRORS - Numbers copied correctly?
    
    For each error found:
    - State the error
    - Explain why it's wrong
    - Provide the correction
    
    If no errors, confirm the solution is correct.
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": detection_prompt}]
    )
    
    error_check = response2.choices[0].message.content
    print("🔍 ERROR DETECTION:")
    print(error_check)
    
    return solution, error_check

# Run the example
solution, errors = error_detection_example()
````

## 🌍 Real-World Example: Code Review

````python
def code_error_detection():
    """Real-world: Detecting errors in generated code."""
    
    # Step 1: Generate code
    code_request = """
    Write a Python function to calculate the average of a list of numbers.
    Include error handling for empty lists.
    """
    
    response1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": code_request}]
    )
    
    code = response1.choices[0].message.content
    print("💻 GENERATED CODE:")
    print(code)
    print("\n" + "="*50 + "\n")
    
    # Step 2: Comprehensive error detection
    review_prompt = f"""
    Review this Python code for errors:
    ```python
    {code}
    ```
    
    Check for these types of errors:
    
    SYNTAX ERRORS:
    - Missing colons, brackets, quotes
    - Indentation issues
    
    LOGIC ERRORS:
    - Incorrect algorithm
    - Edge cases not handled
    - Wrong calculations
    
    RUNTIME ERRORS:
    - Potential exceptions
    - Type mismatches
    - Division by zero
    
    STYLE ISSUES:
    - PEP 8 violations
    - Poor naming
    - Missing docstrings
    
    For each issue found, provide:
    - Issue type and location
    - Why it's a problem
    - Suggested fix
    """
    
    response2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": review_prompt}]
    )
    
    review = response2.choices[0].message.content
    print("🔍 CODE REVIEW RESULTS:")
    print(review)
    
    return code, review

# Run the example
code, review = code_error_detection()
````

## ⚠️ Failure Case: Undetected Errors

````python
def error_detection_failure():
    """Show when error detection can miss problems."""
    
    # Sometimes models miss subtle errors
    flawed_content = """
    The capital of Australia is Sydney. It has a population of
    about 5 million people and is located on the southeastern coast.
    """
    
    print("Content to check:")
    print(flawed_content)
    print()
    
    detection_prompt = f"""
    Check this text for factual errors:
    """
    {flawed_content}
    """
    
    List any errors you find.
    """
    
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": detection_prompt}]
    )
    
    result = response.choices[0].message.content
    print("Error detection result:")
    print(result)
    
    print("\n⚠️ Note: Error detection can miss:")
    print("   - Subtle factual errors (like capital cities)")
    print("   - Errors that sound plausible")
    print("   - Complex logical errors")
    print("   - Context-dependent issues")
    print("\n   (The capital of Australia is actually Canberra, not Sydney)")

error_detection_failure()
````

## 💡 Tips & Tricks

### Model-Specific Advice

**GPT-4/GPT-4o:**
- Excellent at syntax and structural errors
- Good at mathematical error detection
- Use specific error categories for best results

**Claude:**
- Strong at logical error detection
- Good at finding subtle inconsistencies
- Effective for content review

**Gemini:**
- Good for code error detection
- Effective for factual checking
- May need explicit error type instructions

### Best Practices
1. **Be specific about error types** - list what to look for
2. **Use multiple passes** - check different aspects separately
3. **Provide context** - more context helps find subtle errors
4. **Cross-validate** - use external tools for critical checks
5. **Learn from misses** - track what types are often missed
