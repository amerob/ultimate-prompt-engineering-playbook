## 📖 Description
Testing prompt performance across variations in phrasing, formatting, and input style.

**When to use**: Prompt validation before production, ensuring consistent behavior across user populations.

## ⚙️ How It Works
**Step-by-step process for Robustness Testing:**

1. **Prepare** — Understand the task requirements and constraints.
2. **Construct the prompt** — Apply the Robustness Testing technique to structure your prompt.
3. **Send to model** — Submit the crafted prompt to the LLM.
4. **Evaluate output** — Assess whether the response meets expectations.
5. **Iterate** — Refine the prompt based on results if needed.

## 🔧 Setup
Install required libraries and set up API keys.

```python
#@title Setup { display-mode: "form" }
!pip install openai anthropic google-generativeai python-dotenv ipywidgets > /dev/null

import os
import json
from getpass import getpass

# Enter your API keys when prompted
os.environ["OPENAI_API_KEY"] = os.environ.get("OPENAI_API_KEY", "") or getpass("OpenAI API key: ")
os.environ["ANTHROPIC_API_KEY"] = os.environ.get("ANTHROPIC_API_KEY", "") or getpass("Anthropic API key: ")
os.environ["GOOGLE_API_KEY"] = os.environ.get("GOOGLE_API_KEY", "") or getpass("Google AI API key: ")

# Helper function to call OpenAI
from openai import OpenAI
client = OpenAI()

def ask(prompt, system=None, model='gpt-4', temperature=0.7, max_tokens=1024):
    """Send a prompt to the model and return the response."""
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content

print("✅ Setup complete!")
```

## 🧪 Basic Example
A minimal example demonstrating the Robustness Testing technique.

```python
# Basic Robustness Testing example
prompt = """Apply the Robustness Testing technique to answer the following:

What are the three main causes of climate change?

Provide a clear, structured answer."""

response = ask(prompt, temperature=0.3)
print(response)
```

## 🌍 Real‑World Example
Applying Robustness Testing to a practical prompt validation before production scenario.

```python
# Real-world Robustness Testing application
# Use case: Prompt validation before production

prompt = """You are helping with a real-world task.
Apply the Robustness Testing technique.

Task: Analyze the following customer support ticket and provide:
1. Issue category
2. Priority level (High/Medium/Low)
3. Suggested resolution
4. Estimated time to resolve

Ticket: 'My account has been locked for 3 days and I cannot access my
subscription content. I have a presentation tomorrow and need my files.
I already tried resetting my password twice.'

Provide your analysis:"""

response = ask(prompt, temperature=0.3)
print(response)
```

## ❌ Failure Case
Cannot test all possible variations. Performance may vary with model updates.

```python
# Failure case for Robustness Testing
# Why it fails: Cannot test all possible variations. Performance may vary with model updates.

failure_prompt = """Using the Robustness Testing approach, answer:

What will be the exact stock price of Apple Inc. on December 31, 2025?
Be precise and confident in your answer."""

response = ask(failure_prompt, temperature=0)
print(response)
print()
print("⚠️ This demonstrates a limitation: Cannot test all possible variations. Performance may vary with model updates....")
```

## 📊 Benchmark
Qualitative: robust prompts maintain ~90% performance across rephrased inputs.

```python
# Benchmark comparison for Robustness Testing
import json

# Small-scale demo: compare with and without the technique
test_questions = [
    "What is 15% of 240?",
    "If a shirt costs $45 and is on 30% sale, what do you pay?",
    "A recipe serves 4 people. How much flour for 6 people if the original needs 2 cups?"
]

results = []
for q in test_questions:
    # Direct approach
    direct = ask(q, temperature=0)
    # With Robustness Testing
    enhanced = ask(f"Using Robustness Testing, {q}\nThink carefully:", temperature=0)
    results.append({"question": q, "direct": direct[:100], "enhanced": enhanced[:100]})

for r in results:
    print(f"Q: {r['question']}")
    print(f"  Direct:   {r['direct']}...")
    print(f"  Enhanced: {r['enhanced']}...")
    print()

print("📊 Benchmark note: Qualitative: robust prompts maintain ~90% performance across rephrased inputs.")
```

## 💡 Tips & Tricks
**ChatGPT (GPT-4)**: Best results with clear formatting. Experiment with system messages to set up Robustness Testing.

**Claude**: Claude tends to follow instructions precisely. Use XML tags for structure when applying Robustness Testing.

**Gemini**: Works well with explicit step-by-step instructions for Robustness Testing.

**Parameter Tuning**:
- `temperature=0`: Deterministic, best for factual tasks
- `temperature=0.3-0.5`: Balanced creativity and consistency
- `temperature=0.7-1.0`: More creative, diverse outputs

**Common Pitfalls**: Over-engineering the prompt can backfire. Start simple and iterate.
