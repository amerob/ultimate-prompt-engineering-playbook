Code Generation is the practice of using LLMs to write, refactor, debug, and explain code across multiple programming languages and frameworks.

## 📋 Description

Code generation with LLMs involves prompting models to:
- Write new functions, classes, or entire programs
- Refactor and optimize existing code
- Generate unit tests and documentation
- Debug errors and suggest fixes
- Translate code between programming languages
- Explain complex code logic

**Best for**: Rapid prototyping, boilerplate generation, learning new languages, code review assistance

## 🔧 How It Works

```
┌─────────────────────────────────────────────────────────────┐
│  1. SPECIFY: Language, framework, requirements, constraints │
│     ↓                                                       │
│  2. PROVIDE: Context (existing code, data structures)       │
│     ↓                                                       │
│  3. DEFINE: Input/output format, edge cases, examples       │
│     ↓                                                       │
│  4. REQUEST: Code with comments, tests, error handling      │
│     ↓                                                       │
│  5. ITERATE: Review, test, refine                           │
└─────────────────────────────────────────────────────────────┘
```

**Key Elements:**
- Clear specification of programming language and version
- Input/output examples (test cases)
- Constraints (performance, memory, dependencies)
- Style preferences (PEP 8, functional vs OOP)

## 💡 Basic Example

Simple function generation with clear specifications.

````
def generate_code(prompt, model="gpt-4"):
    """Generate code using OpenAI API."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "system",
                "content": "You are an expert software engineer. Write clean, well-documented code with error handling and type hints."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.2,  # Lower for more deterministic code
        max_tokens=1500
    )
    return response.choices[0].message.content

# Basic example: Generate a utility function
basic_prompt = """
Write a Python function that validates email addresses.

Requirements:
- Function name: validate_email
- Input: email string
- Output: tuple (is_valid: bool, error_message: str | None)
- Use regex for validation
- Handle edge cases (empty string, None, etc.)
- Include docstring with examples
"""

print("=== BASIC CODE GENERATION ===")
code = generate_code(basic_prompt)
print(code)
````

## 🌍 Real-World Example

Complete API endpoint generation with database integration.

````
# Real-world: Generate a complete API endpoint
real_world_prompt = """
Create a FastAPI endpoint for a user registration system.

Context:
- Using FastAPI with SQLAlchemy ORM
- PostgreSQL database
- Pydantic for validation

Requirements:
1. POST /api/v1/users/register endpoint
2. Validate: email (unique), password (min 8 chars, 1 uppercase, 1 number), username (3-20 chars)
3. Hash password with bcrypt before storing
4. Return JWT token on success
5. Handle errors: duplicate email, validation errors, database errors
6. Include rate limiting (5 attempts per minute)

Provide:
- Pydantic models (Request/Response)
- SQLAlchemy User model
- The endpoint function
- Error handling middleware
- Unit tests with pytest
"""

print("\n=== REAL-WORLD CODE GENERATION ===")
real_world_code = generate_code(real_world_prompt, model="gpt-4")
print(real_world_code[:3000])  # Print first 3000 chars
print("\n... [truncated for display] ...")
````

## ⚠️ Failure Case

When code generation fails and how to fix it.

````
# Vague prompt that will produce poor results
vague_prompt = "Write a function to process data."

print("=== VAGUE PROMPT (POOR RESULT) ===")
poor_code = generate_code(vague_prompt)
print(poor_code)

print("\n" + "="*50)
print("PROBLEMS WITH VAGUE PROMPTS:")
print("="*50)
print("❌ No language specified")
print("❌ No input/output defined")
print("❌ No error handling requirements")
print("❌ No context about data structure")
print("❌ Generic function name and logic")

print("\n=== IMPROVED PROMPT (BETTER RESULT) ===")
improved_prompt = """
Write a Python function to process CSV data for sales analysis.

Input: pandas DataFrame with columns [date, product_id, quantity, price]
Output: Dictionary with {total_revenue, top_product, daily_sales}

Requirements:
- Calculate total revenue (quantity * price)
- Find top-selling product by quantity
- Group sales by date
- Handle missing values gracefully
- Include type hints and docstring
"""

better_code = generate_code(improved_prompt)
print(better_code)
````

## 💡 Tips & Tricks

### Model-Specific Advice

**GPT-4/GPT-4o:**
- Excellent for complex algorithms and architecture
- Use `temperature=0.2` for consistent results
- Can handle multi-file projects

**Claude 3:**
- Great for detailed explanations alongside code
- Strong at debugging and code review
- Use for educational code examples

**GitHub Copilot:**
- Best for autocomplete within IDE
- Learns from your coding style
- Use comments to guide generation

### Best Practices

1. **Always review generated code** - Never blindly trust LLM output
2. **Test thoroughly** - Run unit tests on all generated code
3. **Check for security** - Review for SQL injection, XSS vulnerabilities
4. **Verify licenses** - Be aware of training data licensing issues
5. **Iterate** - Start simple, then add complexity incrementally
