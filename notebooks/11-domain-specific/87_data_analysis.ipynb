Data Analysis prompting involves using LLMs to explore datasets, generate insights, create visualizations, and perform statistical analysis from data descriptions or sample data.

## 📋 Description

LLM-assisted data analysis enables:
- Exploratory data analysis (EDA) planning
- Statistical test selection and interpretation
- Python/R code for data processing
- Insight generation and pattern recognition
- Visualization recommendations
- Data cleaning and preprocessing strategies
- Report generation and executive summaries

**Best for**: Data scientists, analysts, researchers, business intelligence professionals

## 🔧 How It Works

```
┌─────────────────────────────────────────────────────────────┐
│  1. DATA PROFILE: Columns, types, sample values, size       │
│     ↓                                                       │
│  2. OBJECTIVE: Define analysis goals and questions          │
│     ↓                                                       │
│  3. METHODOLOGY: Statistical tests, ML approaches           │
│     ↓                                                       │
│  4. CODE: Generate analysis code (pandas, sklearn, etc)     │
│     ↓                                                       │
│  5. INTERPRETATION: Explain results in business terms       │
└─────────────────────────────────────────────────────────────┘
```

## 🛠️ Setup

````
# Install required packages
!pip install openai pandas numpy matplotlib seaborn -q

import os
from getpass import getpass
from openai import OpenAI
import pandas as pd
import numpy as np

# Get API key securely
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

print("✓ Setup complete!")
````

## 💡 Basic Example

````
def analyze_data(prompt, model="gpt-4"):
    """Generate data analysis code and insights."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "system",
                "content": "You are an expert data scientist. Provide analysis code, statistical insights, and clear business interpretations."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.3,
        max_tokens=2000
    )
    return response.choices[0].message.content

# Create sample dataset
np.random.seed(42)
sales_data = pd.DataFrame({
    'date': pd.date_range('2023-01-01', periods=100, freq='D'),
    'product': np.random.choice(['A', 'B', 'C'], 100),
    'region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'units_sold': np.random.randint(10, 100, 100),
    'revenue': np.random.randint(1000, 10000, 100)
})

print("Sample Data:")
print(sales_data.head())
print(f"\nDataset shape: {sales_data.shape}")

# Basic analysis prompt
basic_prompt = f"""
Analyze this sales dataset:

Columns: {list(sales_data.columns)}
Shape: {sales_data.shape}
Sample data:
{sales_data.head(10).to_string()}

Provide:
1. Summary statistics for numeric columns
2. Distribution analysis by product and region
3. Python code to generate these insights
4. Key observations and recommendations
"""

print("\n=== BASIC DATA ANALYSIS ===")
analysis = analyze_data(basic_prompt)
print(analysis)
````

## 🌍 Real-World Example

````
# Real-world: Customer churn analysis
churn_data = pd.DataFrame({
    'customer_id': range(1, 101),
    'tenure_months': np.random.randint(1, 72, 100),
    'monthly_charges': np.random.uniform(20, 120, 100).round(2),
    'total_charges': np.random.uniform(100, 8000, 100).round(2),
    'contract_type': np.random.choice(['Month-to-month', 'One year', 'Two year'], 100, p=[0.5, 0.3, 0.2]),
    'payment_method': np.random.choice(['Electronic check', 'Mailed check', 'Bank transfer', 'Credit card'], 100),
    'internet_service': np.random.choice(['DSL', 'Fiber optic', 'No'], 100, p=[0.4, 0.4, 0.2]),
    'churn': np.random.choice([0, 1], 100, p=[0.7, 0.3])
})

real_world_prompt = f"""
Customer Churn Analysis Dataset:

Columns and descriptions:
- customer_id: Unique identifier
- tenure_months: How long customer has been with company
- monthly_charges: Monthly subscription cost
- total_charges: Total amount paid (tenure * monthly_charges)
- contract_type: Month-to-month, One year, Two year
- payment_method: Payment method used
- internet_service: Type of internet service
- churn: 1 = churned, 0 = retained

Sample data (10 rows):
{churn_data.head(10).to_string()}

Churn rate: {churn_data['churn'].mean():.1%}

Perform comprehensive churn analysis:
1. Identify key factors correlated with churn
2. Segment customers by risk level
3. Recommend statistical tests to validate findings
4. Provide actionable business recommendations
5. Suggest visualizations for executive presentation
"""

print("\n=== CHURN ANALYSIS ===")
churn_analysis = analyze_data(real_world_prompt)
print(churn_analysis[:4000])
print("\n... [truncated] ...")
````

## 💡 Tips & Tricks

### Data Analysis Prompting Best Practices

1. **Provide sample data** - Include 5-10 representative rows
2. **Specify data types** - Numeric, categorical, datetime, etc.
3. **Mention data quality issues** - Missing values, outliers, duplicates
4. **Define success metrics** - What constitutes a good insight?
5. **Request visualizations** - Charts make insights more actionable

### Common Analysis Patterns

- **Descriptive Analysis**: Summary statistics, distributions, correlations
- **Diagnostic Analysis**: Root causes, segmentation, anomaly detection
- **Predictive Analysis**: Forecasting, churn prediction, trend analysis
- **Prescriptive Analysis**: Recommendations, A/B tests, optimization

### Libraries to Mention
- **pandas**: Data manipulation
- **numpy**: Numerical operations
- **matplotlib/seaborn/plotly**: Visualization
- **scipy**: Statistical tests
- **sklearn**: Machine learning
- **statsmodels**: Statistical modeling
