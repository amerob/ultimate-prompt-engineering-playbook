Technical Documentation prompting involves generating clear, structured documentation for APIs, codebases, systems, and processes that serves both technical and non-technical audiences.

## 📋 Description

LLM-assisted technical documentation includes:
- API documentation (endpoints, parameters, responses)
- Code documentation (docstrings, comments, READMEs)
- Architecture and system design docs
- User guides and tutorials
- Troubleshooting guides
- Release notes and changelogs
- Process documentation

**Best for**: Technical writers, developers, DevOps engineers, product managers

## 🔧 How It Works

```
┌─────────────────────────────────────────────────────────────┐
│  1. AUDIENCE: Define who will read the documentation        │
│     ↓                                                       │
│  2. SCOPE: What to cover and at what depth                  │
│     ↓                                                       │
│  3. SOURCE: Code, specs, or subject matter input            │
│     ↓                                                       │
│  4. FORMAT: Structure, style guide, templates               │
│     ↓                                                       │
│  5. OUTPUT: Generate draft with examples and diagrams       │
│     ↓                                                       │
│  6. REVIEW: Technical accuracy and completeness             │
└─────────────────────────────────────────────────────────────┘
```

## 💡 Basic Example

````
def generate_docs(prompt, model="gpt-4"):
    """Generate technical documentation using OpenAI API."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "system",
                "content": "You are a technical documentation expert. Write clear, accurate documentation following industry best practices."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.3,
        max_tokens=2000
    )
    return response.choices[0].message.content

# Basic example: Function documentation
sample_code = '''
def process_payment(amount, currency, customer_id, payment_method):
    # Validate inputs
    if amount <= 0:
        raise ValueError("Amount must be positive")
    
    # Convert currency if needed
    if currency != 'USD':
        amount = convert_currency(amount, currency, 'USD')
    
    # Process payment
    result = payment_gateway.charge(customer_id, amount, payment_method)
    
    # Log transaction
    log_transaction(customer_id, amount, result.transaction_id)
    
    return result
'''

basic_prompt = f"""
Generate comprehensive documentation for this Python function:

```python
{sample_code}
```

Include:
1. Function description
2. Parameter table (name, type, required, description)
3. Return value description
4. Exceptions that can be raised
5. Usage example
6. Notes about currency conversion
"""

print("=== BASIC DOCUMENTATION ===")
docs = generate_docs(basic_prompt)
print(docs)
````

## 🌍 Real-World Example

````
# Real-world: Complete API documentation
api_spec = """
API: E-Commerce Order Management Service
Base URL: https://api.example.com/v1
Authentication: Bearer token in Authorization header

Endpoints:

1. GET /orders
   - Query params: status, start_date, end_date, page, limit
   - Returns: List of orders with pagination
   - Rate limit: 100 req/min

2. POST /orders
   - Body: customer_id, items[], shipping_address, payment_method
   - Returns: Created order with order_id
   - Errors: 400 (validation), 402 (payment failed), 409 (inventory)

3. GET /orders/{order_id}
   - Returns: Full order details including status history
   - Errors: 404 (not found)

4. PATCH /orders/{order_id}/status
   - Body: status, reason
   - Valid transitions: pending→processing→shipped→delivered
   - Errors: 400 (invalid transition), 403 (unauthorized)

5. DELETE /orders/{order_id}
   - Only allowed for pending orders
   - Soft delete (archived)
"""

real_world_prompt = f"""
Create complete API documentation for this e-commerce service:

{api_spec}

Format: OpenAPI 3.0 style documentation

Include for each endpoint:
1. Description and use case
2. Authentication requirements
3. Request parameters (path, query, body)
4. Response schemas with examples
5. Error responses with codes and messages
6. Code examples in Python, cURL, and JavaScript

Also include:
- Overview section
- Authentication guide
- Rate limiting details
- Common error codes reference
"""

print("\n=== API DOCUMENTATION ===")
api_docs = generate_docs(real_world_prompt)
print(api_docs[:4000])
print("\n... [truncated] ...")
````

## ⚠️ Failure Case

````
# Insufficient context leads to generic docs
poor_prompt = "Document my API."

print("=== POOR PROMPT ===")
print("This will fail because:")
print("❌ No API endpoints specified")
print("❌ No authentication details")
print("❌ No examples or use cases")
print("❌ No target audience defined")
print("❌ No format or style specified")

print("\n=== IMPROVED APPROACH ===")
improved_prompt = """
Create API documentation for a payment processing service.

Audience: Backend developers integrating the API
Format: Markdown with OpenAPI 3.0 structure

Key Endpoints:
1. POST /payments - Create a new payment
2. GET /payments/{id} - Retrieve payment status
3. POST /payments/{id}/refund - Process a refund

Authentication: API key in X-API-Key header
Base URL: https://payments.api.com/v2

Include:
- Request/response examples
- Error handling
- Webhook specifications
- Rate limiting information
"""

better_docs = generate_docs(improved_prompt)
print(better_docs[:2000])
print("\n... [truncated] ...")
````

## 💡 Tips & Tricks

### Documentation Best Practices

1. **Know your audience** - Adjust technical depth accordingly
2. **Start with overview** - Big picture before details
3. **Use examples liberally** - Show, don't just tell
4. **Include troubleshooting** - Anticipate common issues
5. **Keep it current** - Documentation decays quickly

### Prompt Templates

**API Documentation:**
```
Create API documentation for [service name].

Endpoints:
- [METHOD] [path] - [description]

Authentication: [method]
Include: Examples, error codes, rate limits
Format: [OpenAPI/Markdown/etc]
```

**Code Documentation:**
```
Document this [language] code:
[code]

Include:
- Function/class purpose
- Parameters and return values
- Usage examples
- Edge cases and limitations
```

**Tutorial:**
```
Create a tutorial for [task] using [technology].

Prerequisites: [list]
Difficulty: [beginner/intermediate/advanced]

Include:
- Step-by-step instructions
- Code snippets
- Expected output
- Common pitfalls
```
