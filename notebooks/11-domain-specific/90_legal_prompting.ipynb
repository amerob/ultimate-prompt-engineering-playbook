Legal Prompting involves using LLMs for legal document analysis, contract review, legal research assistance, and compliance checking. **Note: LLMs are not a substitute for qualified legal counsel.**

## 📋 Description

LLM-assisted legal tasks include:
- Contract clause extraction and analysis
- Legal document summarization
- Compliance checklist generation
- Legal research assistance
- Risk identification in agreements
- Document comparison and redlining
- Regulatory requirement mapping

**⚠️ IMPORTANT DISCLAIMER:**
- LLMs cannot provide legal advice
- Always consult qualified legal professionals
- Verify all AI-generated legal content
- Be aware of jurisdiction-specific requirements

**Best for**: Legal professionals as a research and drafting aid only

## 🔧 How It Works

```
┌─────────────────────────────────────────────────────────────┐
│  1. DOCUMENT: Provide the legal text to analyze             │
│     ↓                                                       │
│  2. CONTEXT: Jurisdiction, document type, purpose           │
│     ↓                                                       │
│  3. TASK: Specific analysis or extraction goal              │
│     ↓                                                       │
│  4. CONSTRAINTS: What to flag, highlight, or ignore         │
│     ↓                                                       │
│  5. OUTPUT: Structured analysis with citations              │
└─────────────────────────────────────────────────────────────┘
```

## 💡 Basic Example

````
def legal_analysis(prompt, model="gpt-4"):
    """Generate legal document analysis using OpenAI API."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "system",
                "content": "You are a legal document analysis assistant. Provide structured analysis while emphasizing that this is not legal advice. Always recommend consulting a qualified attorney."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.2,
        max_tokens=2000
    )
    return response.choices[0].message.content

# Basic example: Contract clause extraction
sample_contract = """
SERVICE AGREEMENT

This Agreement is entered into between Company A ("Provider") and Company B ("Client")
effective January 1, 2024.

1. TERM: This agreement shall commence on January 1, 2024 and continue for a period
of twelve (12) months, automatically renewing for successive one-year periods unless
terminated in accordance with Section 8.

2. SERVICES: Provider shall deliver cloud hosting services as described in Exhibit A.

3. PAYMENT: Client shall pay monthly fees as specified in Exhibit B. Late payments
subject to 1.5% monthly service charge.

4. LIMITATION OF LIABILITY: Provider's total liability shall not exceed the total
amount paid by Client in the twelve months preceding the claim.

5. INDEMNIFICATION: Client agrees to indemnify Provider against third-party claims
arising from Client's use of the services.

6. DATA SECURITY: Provider shall maintain industry-standard security measures and
notify Client of any data breach within 72 hours.

7. TERMINATION FOR CONVENIENCE: Either party may terminate with ninety (90) days
written notice.

8. TERMINATION FOR CAUSE: Either party may terminate immediately for material breach
if uncured within thirty (30) days of notice.
"""

basic_prompt = f"""
Analyze this service agreement and extract:

```
{sample_contract}
```

1. Key dates and deadlines
2. Financial obligations and penalties
3. Termination conditions
4. Liability limitations
5. Data protection provisions
6. Any clauses that might warrant legal review

Present findings in a structured table format.
"""

print("=== BASIC LEGAL ANALYSIS ===")
analysis = legal_analysis(basic_prompt)
print(analysis)
````

## 🌍 Real-World Example

````
# Real-world: Multi-document compliance review
real_world_prompt = """
You are conducting a preliminary review for GDPR compliance.

Document to analyze (Privacy Policy excerpt):

---
PRIVACY POLICY

Data Collection: We collect personal information including name, email, phone number,
and browsing behavior through cookies and tracking pixels.

Data Use: We use your data to provide services, send marketing communications,
and share with selected partners for advertising purposes.

Data Retention: We retain personal data for as long as necessary for business purposes
or as required by law.

Your Rights: You may contact us to update your information.

Cookies: By using our site, you consent to our use of cookies.
---

Task: Conduct a GDPR compliance gap analysis covering:

1. Lawful basis for processing (Article 6)
2. Data subject rights (Articles 15-22)
3. Consent requirements (Article 7)
4. Privacy by design considerations
5. International transfer safeguards
6. Data breach notification procedures

For each area, indicate:
- Compliant (✓), Partial (△), or Non-compliant (✗)
- Specific gaps or concerns
- Recommendations for improvement

Note: This is for preliminary review only. Formal legal review required.
"""

print("\n=== GDPR COMPLIANCE ANALYSIS ===")
compliance_analysis = legal_analysis(real_world_prompt)
print(compliance_analysis[:4000])
print("\n... [truncated] ...")
````

## 💡 Tips & Tricks

### Legal Prompting Best Practices

1. **Always include disclaimers** - Make it clear this is not legal advice
2. **Specify jurisdiction** - Laws vary significantly by location
3. **Request structured output** - Tables, checklists, numbered lists
4. **Ask for citations** - Request reference to specific laws or sections
5. **Use for preliminary review only** - Always follow up with professionals

### What LLMs Can Help With

✅ **Appropriate Uses:**
- Document organization and summarization
- Clause extraction and categorization
- Creating checklists and templates
- Explaining general legal concepts
- Translating legal jargon to plain language
- Initial compliance gap identification

❌ **Never Use For:**
- Specific legal advice for your situation
- Drafting binding legal documents without review
- Interpreting laws for your jurisdiction
- Making litigation decisions
- Compliance determinations

### Prompt Templates

**Document Review:**
```
Review this [document type] for [purpose].

Jurisdiction: [location]
Document: [text]

Extract/Analyze:
- [specific items]

Note: This is for preliminary review only.
Not a substitute for legal counsel.
```

**Compliance Check:**
```
Review this document for [regulation] compliance.

Regulation: [name and version]
Jurisdiction: [applicable regions]
Document: [text]

Identify gaps and recommend improvements.
```
