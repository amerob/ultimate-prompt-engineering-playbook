Marketing Copy prompting involves creating compelling, persuasive content for advertising, brand messaging, product descriptions, and various marketing channels.

## 📋 Description

LLM-assisted marketing copywriting includes:
- Ad copy for various platforms (Google, Facebook, LinkedIn)
- Product descriptions and sales pages
- Email marketing campaigns
- Social media content
- Brand voice and messaging guidelines
- Landing page copy
- Video scripts and storyboards
- SEO-optimized content

**Best for**: Marketers, copywriters, content creators, entrepreneurs, agencies

## 🔧 How It Works

```
┌─────────────────────────────────────────────────────────────┐
│  1. BRAND: Voice, tone, values, personality                 │
│     ↓                                                       │
│  2. AUDIENCE: Target demographics, pain points, desires     │
│     ↓                                                       │
│  3. PRODUCT: Features, benefits, unique value proposition   │
│     ↓                                                       │
│  4. CHANNEL: Platform constraints and best practices        │
│     ↓                                                       │
│  5. OBJECTIVE: Awareness, conversion, engagement, etc.      │
│     ↓                                                       │
│  6. CTA: Clear call-to-action and next steps                │
└─────────────────────────────────────────────────────────────┘
```

## 🛠️ Setup

````
# Install required packages
!pip install openai -q

import os
from getpass import getpass
from openai import OpenAI

# Get API key securely
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

print("✓ Setup complete!")
````

## 💡 Basic Example

````
def generate_copy(prompt, model="gpt-4", temperature=0.7):
    """Generate marketing copy using OpenAI API."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "system",
                "content": "You are an expert copywriter. Create compelling, persuasive marketing copy that drives action while maintaining brand authenticity."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=temperature,
        max_tokens=1500
    )
    return response.choices[0].message.content

# Basic example: Product description
basic_prompt = """
Write a compelling product description for a sustainable water bottle.

Product details:
- Name: EcoFlow Bottle
- Material: 100% recycled stainless steel
- Features: Keeps drinks cold 24h, hot 12h, leak-proof
- Price: $35
- Target: Environmentally conscious millennials

Tone: Inspiring, eco-friendly, premium but accessible
Length: 100-150 words
Include: Headline, body copy, and call-to-action
"""

print("=== BASIC MARKETING COPY ===")
copy = generate_copy(basic_prompt)
print(copy)
````

## 🌍 Real-World Example

````
# Real-world: Multi-channel campaign
real_world_prompt = """
Create a complete marketing campaign for a new fitness app called "FitPulse".

Product Overview:
- AI-powered personalized workout plans
- Real-time form correction via camera
- Nutrition tracking and meal suggestions
- Community challenges and leaderboards
- Price: $14.99/month or $99/year

Target Audience:
- Primary: 25-40 year olds, busy professionals
- Pain points: No time for gym, need guidance, want results
- Motivations: Convenience, personalization, community

Brand Voice:
- Energetic but not aggressive
- Supportive and inclusive
- Science-backed but accessible

Create copy for:
1. Facebook ad (primary text + headline + description)
2. Google search ad (3 headlines + 2 descriptions)
3. Email subject line + preview text + opening
4. Landing page hero section
5. Instagram caption for launch post
6. App store description (short + long)

Each piece should be optimized for its platform and include appropriate CTAs.
"""

print("\n=== MULTI-CHANNEL CAMPAIGN ===")
campaign = generate_copy(real_world_prompt, temperature=0.8)
print(campaign[:4000])
print("\n... [truncated] ...")
````

### Marketing Prompt Best Practices

1. **Define your avatar** - Specific demographics and psychographics
2. **Lead with benefits** - Not features
3. **Use emotional triggers** - Fear, aspiration, belonging
4. **Include social proof** - Numbers, testimonials, authority
5. **Create urgency** - Limited time, scarcity
6. **Test variations** - A/B test different approaches

### Prompt Templates

**Product Description:**
```
Write a [tone] product description for [product name].

Key features:
- [feature 1]: [benefit]

Target audience: [description]
Price point: [range]
Length: [word count]
Include: [specific elements]
```

**Ad Copy:**
```
Write [platform] ad copy for [product/service].

Objective: [awareness/leads/sales]
Target: [audience description]
Key message: [main point]
CTA: [desired action]
Tone: [brand voice]
```

**Email Sequence:**
```
Write a [number]-email sequence for [purpose].

Trigger: [what starts the sequence]
Goal: [desired outcome]
Timeline: [send schedule]
Tone: [brand voice]
```
