Meta-prompting is the technique of using prompts to generate, refine, or improve other prompts. It treats prompt engineering itself as a task that can be delegated to AI.

## Description

Meta-prompting involves asking an AI to:
- Generate prompts for specific tasks
- Improve existing prompts
- Analyze prompt effectiveness
- Create prompt variations
- Debug problematic prompts

**When to use:**
- When you need a high-quality prompt quickly
- When you're stuck on prompt formulation
- When optimizing prompts for specific models
- When creating prompt libraries
- When teaching prompt engineering to others

## How It Works

```
┌─────────────────────────────────────────────────────────────┐
│                    META-PROMPTING FLOW                      │
└─────────────────────────────────────────────────────────────┘

  Step 1: Define Your Goal
         ↓
  Step 2: Create Meta-Prompt
         ↓
  Step 3: Generate Target Prompt
         ↓
  Step 4: Test & Iterate
         ↓
  Step 5: Deploy Optimized Prompt

Example Meta-Prompt Structure:
┌────────────────────────────────────────────────────────────┐
│ "Create a prompt that helps [AI] to [TASK] for [AUDIENCE] │
│  The prompt should include:                                │
│  - [Specific requirements]                                 │
│  - [Output format]                                         │
│  - [Constraints]"                                          │
└────────────────────────────────────────────────────────────┘
```

## Basic Example: Simple Meta-Prompt

Let's create a meta-prompt to generate a customer service response prompt.

````python
def generate_prompt(meta_prompt, model="gpt-4o"):
    """Generate a prompt using meta-prompting."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "You are an expert prompt engineer. Create high-quality, detailed prompts based on the user's request."},
            {"role": "user", "content": meta_prompt}
        ],
        temperature=0.7
    )
    return response.choices[0].message.content

# Simple meta-prompt example
simple_meta = """
Create a prompt that helps an AI write empathetic customer service responses.
The prompt should:
- Ensure responses acknowledge customer frustration
- Include a structure for responses
- Specify tone guidelines
"""

generated_prompt = generate_prompt(simple_meta)
print("=== GENERATED PROMPT ===")
print(generated_prompt)
````

## Real-World Example: Complete Meta-Prompting Workflow

A comprehensive example showing the full meta-prompting pipeline for creating a code review prompt.

````python
# Step 1: Create detailed meta-prompt
detailed_meta = """
Create a comprehensive prompt for AI-assisted code review that:

TASK: Review Python code for best practices, bugs, and improvements

REQUIREMENTS:
1. Check for PEP 8 compliance
2. Identify potential bugs or edge cases
3. Suggest performance optimizations
4. Evaluate code readability and maintainability
5. Check for security vulnerabilities

OUTPUT FORMAT:
- Summary (2-3 sentences)
- Issues found (categorized by severity: Critical/Warning/Suggestion)
- Specific recommendations with code examples
- Overall code quality score (1-10)

CONSTRAINTS:
- Be constructive and educational
- Provide actionable feedback
- Include references to best practices
"""

# Generate the code review prompt
code_review_prompt = generate_prompt(detailed_meta)
print("=== GENERATED CODE REVIEW PROMPT ===")
print(code_review_prompt)
print("\n" + "="*50 + "\n")

# Step 2: Test the generated prompt with sample code
sample_code = """
def calculate_average(numbers):
    total = 0
    for n in numbers:
        total += n
    return total / len(numbers)
"""

print("=== TESTING GENERATED PROMPT ===")
test_response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": code_review_prompt},
        {"role": "user", "content": f"Please review this code:\n\n```python\n{sample_code}\n```"}
    ],
    temperature=0.3
)
print(test_response.choices[0].message.content)
````

## Failure Case: When Meta-Prompting Fails

Meta-prompting can fail when the meta-prompt itself is vague or contradictory.

````python
# Poor meta-prompt example
poor_meta = "Make a good prompt for writing stuff."

print("=== POOR META-PROMPT ===")
print(poor_meta)
print("\n=== RESULT (Likely too generic) ===")

poor_result = generate_prompt(poor_meta)
print(poor_result)

print("\n" + "="*50)
print("PROBLEM: The meta-prompt lacks:")
print("- Specific task definition")
print("- Target audience")
print("- Output requirements")
print("- Constraints or guidelines")
print("\nFIX: Add specific details about what 'writing stuff' means.")
````

## Tips & Tricks

### Model-Specific Advice

| Model | Best Practices |
|-------|---------------|
| GPT-4o | Use structured meta-prompts with clear sections |
| Claude | Include examples of desired output format |
| Gemini | Specify context and constraints explicitly |

### Best Practices

1. **Be Specific**: Include exact requirements, not vague descriptions
2. **Use Structure**: Organize meta-prompts with clear sections
3. **Provide Examples**: Show what good output looks like
4. **Iterate**: Test generated prompts and refine meta-prompts
5. **Document**: Keep a library of effective meta-prompts

### Common Patterns

```
Pattern 1: Task + Format + Constraints
Pattern 2: Role + Task + Examples
Pattern 3: Context + Requirements + Evaluation Criteria
```
