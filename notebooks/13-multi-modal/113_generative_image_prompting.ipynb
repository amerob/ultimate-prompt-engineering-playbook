## 1. Description

**Generative Image Prompting** is the specialized skill of crafting text descriptions that guide AI image generation models to produce desired visual outputs. Unlike standard text prompting, image prompting requires understanding visual composition, artistic terminology, and the unique syntax preferences of different image generation platforms.

### When to Use:
- Creating custom illustrations, artwork, or graphics
- Generating concept art for design projects
- Producing marketing visuals and social media content
- Visualizing ideas before production
- Creating unique stock imagery alternatives
- Exploring artistic styles and visual concepts

### Supported Platforms:
| Platform | Model | Best For |
|----------|-------|----------|
| OpenAI | DALL-E 3 | Photorealism, text rendering |
| Midjourney | v6 | Artistic, stylized imagery |
| Stability AI | Stable Diffusion XL | Customization, local deployment |
| Adobe | Firefly | Commercial-safe content |
| Google | Imagen 2 | High-quality photorealism |

## 2. How It Works

### The Anatomy of an Effective Image Prompt

```
[Subject] + [Action/Context] + [Environment] + [Style] + [Technical Specs] + [Quality Boosters]
```

### Prompt Structure Flow

```
┌─────────────────────────────────────────────────────────────┐
│                    IMAGE PROMPT STRUCTURE                    │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│  ┌──────────────┐    ┌──────────────┐    ┌──────────────┐  │
│  │   SUBJECT    │───▶│   CONTEXT    │───▶│   STYLE      │  │
│  │  (What)      │    │  (Where/When)│    │  (How)       │  │
│  └──────────────┘    └──────────────┘    └──────────────┘  │
│         │                   │                   │          │
│         ▼                   ▼                   ▼          │
│  ┌──────────────────────────────────────────────────────┐ │
│  │              TECHNICAL SPECIFICATIONS                 │ │
│  │  • Camera angle    • Lighting    • Resolution        │ │
│  │  • Lens type       • Color grade  • Aspect ratio     │ │
│  └──────────────────────────────────────────────────────┘ │
│                            │                               │
│                            ▼                               │
│  ┌──────────────────────────────────────────────────────┐ │
│  │              QUALITY MODIFIERS                        │ │
│  │  'highly detailed' '8k' 'masterpiece' 'professional'  │ │
│  └──────────────────────────────────────────────────────┘ │
│                            │                               │
│                            ▼                               │
│              ┌─────────────────────┐                       │
│              │   GENERATED IMAGE   │                       │
│              └─────────────────────┘                       │
│                                                             │
└─────────────────────────────────────────────────────────────┘
```

### Platform-Specific Syntax

**DALL-E 3 (Conversational):**
- Natural language works best
- Can handle complex, detailed descriptions
- Good at following specific instructions

**Midjourney (Structured):**
- Uses parameters like `--ar 16:9`, `--v 6`, `--style raw`
- Weight syntax: `::2` for emphasis
- Image prompts with URLs

**Stable Diffusion (Token-based):**
- Comma-separated keywords effective
- Negative prompts crucial
- Weight syntax: `(word:1.5)` for emphasis

## 4. Basic Example

### Simple Image Generation with DALL-E 3

Let's start with a basic prompt and see how different structures affect output.

````python
def generate_image(prompt, size="1024x1024", model="dall-e-3"):
    """Generate image using DALL-E and display it."""
    try:
        response = client.images.generate(
            model=model,
            prompt=prompt,
            size=size,
            quality="standard",
            n=1,
        )
        
        # Get image URL
        image_url = response.data[0].url
        
        # Download and display
        img_response = requests.get(image_url)
        img = Image.open(BytesIO(img_response.content))
        
        print(f"Prompt: {prompt}")
        print(f"Revised: {response.data[0].revised_prompt if hasattr(response.data[0], 'revised_prompt') else 'N/A'}")
        display(img)
        
        return image_url
    except Exception as e:
        print(f"Error: {e}")
        return None

# Basic prompt
basic_prompt = "A cat sitting on a couch"
print("=== BASIC PROMPT ===")
# generate_image(basic_prompt)  # Uncomment to run
````

````python
# Enhanced prompt with structure
enhanced_prompt = """
A fluffy orange tabby cat with green eyes sitting regally on a vintage velvet couch,
soft natural window lighting, cozy living room interior, photorealistic style,
highly detailed fur texture, professional pet photography, 8k quality
""".strip()

print("=== ENHANCED PROMPT ===")
print(f"Prompt length: {len(enhanced_prompt)} characters")
print(f"Prompt:\n{enhanced_prompt}\n")
# generate_image(enhanced_prompt)  # Uncomment to run
````

## 6. Failure Case

### Common Mistakes and How to Fix Them

#### ❌ Failure 1: Vague Descriptions

````python
bad_prompt_1 = "Make a nice picture of something"

good_prompt_1 = """
A serene Japanese zen garden with raked sand patterns,
a single red maple tree, morning mist,
minimalist composition, peaceful atmosphere,
fine art photography style
""".strip()

print("❌ BAD:", bad_prompt_1)
print("\n✅ GOOD:", good_prompt_1)
````

**Problem:** Too vague - AI doesn't know what "nice" means or what "something" is.  
**Fix:** Be specific about subject, style, and mood.

#### ❌ Failure 2: Conflicting Instructions

````python
bad_prompt_2 = "A photorealistic cartoon character in a realistic anime style"

good_prompt_2a = """
A 3D rendered Pixar-style character,
cartoon proportions, vibrant colors,
soft studio lighting, family-friendly design
""".strip()

good_prompt_2b = """
A photorealistic portrait of a person,
studio photography, dramatic lighting,
highly detailed skin texture, professional headshot
""".strip()

print("❌ BAD (conflicting):", bad_prompt_2)
print("\n✅ GOOD (cartoon):", good_prompt_2a)
print("\n✅ GOOD (realistic):", good_prompt_2b)
````

**Problem:** "Photorealistic" and "cartoon" are contradictory.  
**Fix:** Choose one clear style direction.

#### ❌ Failure 3: Overloading the Prompt

````python
bad_prompt_3 = """
A dragon flying over a castle with knights fighting below and a princess
watching from a tower and a wizard casting a spell and a storm with lightning
and fire everywhere and a full moon and stars and a rainbow and unicorns
and a spaceship and aliens and dinosaurs
""".strip()

good_prompt_3 = """
A majestic dragon soaring above a medieval castle at sunset,
golden light illuminating stone towers, dramatic clouds,
epic fantasy illustration style, cinematic composition
""".strip()

print("❌ BAD (overloaded):")
print(bad_prompt_3)
print("\n✅ GOOD (focused):")
print(good_prompt_3)
````

**Problem:** Too many elements confuse the model and produce incoherent results.  
**Fix:** Focus on 1-3 main elements with clear relationships.

## 9. Tips & Tricks

### Platform-Specific Best Practices

#### DALL-E 3 Tips
- ✅ Use natural, conversational language
- ✅ Be specific about composition and layout
- ✅ Specify text content if needed (excellent text rendering)
- ❌ Avoid excessive repetition of keywords
- ❌ Don't use parameter syntax (e.g., `--ar 16:9`)

#### Midjourney Tips
- ✅ Use comma-separated descriptive phrases
- ✅ Utilize parameters: `--ar`, `--v`, `--style`, `--s`
- ✅ Use image prompts for style transfer
- ✅ Apply weights with `::` for emphasis
- ✅ Use `--no` for negative prompting

#### Stable Diffusion Tips
- ✅ Use token weighting: `(word:1.5)` for emphasis
- ✅ Always include negative prompts
- ✅ Use specific LoRA models for styles
- ✅ Experiment with different samplers
- ✅ Use ControlNet for pose/structure control

### Universal Best Practices

1. **Start with the subject** - What is the main focus?
2. **Add context** - Where/when is this happening?
3. **Specify style** - How should it look?
4. **Include quality terms** - Resolution, detail level
5. **Iterate and refine** - Generate, evaluate, improve

### Effective Quality Modifiers

| Category | Keywords |
|----------|----------|
| Resolution | 4k, 8k, ultra HD, high resolution |
| Detail | highly detailed, intricate, fine details |
| Professional | professional photography, award-winning, masterpiece |
| Rendering | octane render, unreal engine, ray tracing |
| Artistic | trending on ArtStation, digital art, concept art |
